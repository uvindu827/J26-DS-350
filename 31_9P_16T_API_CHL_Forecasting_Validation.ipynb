```python
# ======================================================================
# STEP 31.9P-16T-23-C-FIX
# COPERNICUS LOGIN & SURFACE-LEVEL VERIFICATION
# ======================================================================

print("=" * 70)
print("STEP 31.9P-16T-23-C-FIX")
print("COPERNICUS LOGIN & SURFACE-LEVEL VERIFICATION")
print("=" * 70)


# ======================================================================
# 1. IMPORTS
# ======================================================================

from pathlib import Path
import json
import inspect
import copernicusmarine


# ======================================================================
# 2. PROJECT PATHS
# ======================================================================

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

EXTERNAL_VALIDATION_DIR = (
    PROJECT_DIR
    / "06_models/02_XGBoost/"
      "hyperparameter_optimization/"
      "final_tuned_candidate/"
      "external_validation"
)

EXTERNAL_VALIDATION_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUTPUT_FILE = (
    EXTERNAL_VALIDATION_DIR
    / "copernicus_2026_surface_verification.json"
)


# ======================================================================
# 3. VALIDATION CONFIGURATION
# ======================================================================

VALIDATION_START = "2026-01-01T00:00:00"
VALIDATION_END   = "2026-10-31T23:59:59"

MIN_LATITUDE = 5.8333335
MAX_LATITUDE = 6.9166665

MIN_LONGITUDE = 79.333336
MAX_LONGITUDE = 80.0


# ======================================================================
# 4. DATASET DEFINITIONS
# ======================================================================

DATASETS = {

    "CHL": {
        "dataset_id":
            "cmems_obs-oc_glo_bgc-plankton_my_l3-multi-4km_P1D",
        "expected_variables": [
            "chlorophyll",
            "chlorophyll_a",
            "chl",
            "CHL"
        ]
    },

    "SST": {
        "dataset_id":
            "cmems_mod_glo_phy-thetao_anfc_0.083deg_P1D-m",
        "expected_variables": [
            "thetao",
            "temperature"
        ]
    },

    "SSS": {
        "dataset_id":
            "cmems_mod_glo_phy-so_anfc_0.083deg_P1D-m",
        "expected_variables": [
            "so",
            "salinity"
        ]
    },

    "NUTRIENTS": {
        "dataset_id":
            "cmems_mod_glo_bgc-nut_anfc_0.25deg_P1D-m",
        "expected_variables": [
            "no3",
            "nitrate",
            "po4",
            "phosphate",
            "si",
            "silicate"
        ]
    }
}


# ======================================================================
# 5. MODEL SAFETY CHECK
# ======================================================================

print("\n" + "=" * 70)
print("FINAL MODEL SAFETY CHECK")
print("=" * 70)

FINAL_MODEL = (
    PROJECT_DIR
    / "06_models/02_XGBoost/"
      "hyperparameter_optimization/"
      "final_tuned_candidate/"
      "xgboost_final_tuned_candidate.json"
)

FINAL_FEATURES = (
    PROJECT_DIR
    / "05_model_preprocessing/"
      "final_model_feature_columns.csv"
)

SHAP_DIR = (
    PROJECT_DIR
    / "06_models/02_XGBoost/"
      "hyperparameter_optimization/"
      "final_tuned_candidate/"
      "shap_explainability"
)

THREE_D_DIR = (
    PROJECT_DIR
    / "06_models/02_XGBoost/"
      "hyperparameter_optimization/"
      "final_tuned_candidate/"
      "3d_spatial_temporal_visualization"
)

print(
    "Final tuned XGBoost :",
    FINAL_MODEL.exists()
)

print(
    "Final feature list  :",
    FINAL_FEATURES.exists()
)

print(
    "SHAP directory      :",
    SHAP_DIR.exists()
)

print(
    "3D directory        :",
    THREE_D_DIR.exists()
)

assert FINAL_MODEL.exists()
assert FINAL_FEATURES.exists()

print("\n✓ Final model artifacts remain untouched.")


# ======================================================================
# 6. CHECK TOOLBOX VERSION
# ======================================================================

print("\n" + "=" * 70)
print("COPERNICUS MARINE TOOLBOX")
print("=" * 70)

print(
    "Installed version:",
    getattr(
        copernicusmarine,
        "__version__",
        "unknown"
    )
)

print(
    "\nLogin function signature:"
)

print(
    inspect.signature(
        copernicusmarine.login
    )
)


# ======================================================================
# 7. COPERNICUS LOGIN
# ======================================================================

print("\n" + "=" * 70)
print("COPERNICUS MARINE LOGIN")
print("=" * 70)

print(
    "\nThe official Toolbox login function will be used."
)

print(
    "Your password will not be written into the "
    "verification JSON."
)

print(
    "\nA Copernicus login prompt will appear below."
)

try:

    login_result = copernicusmarine.login()

    print(
        "\n✓ Copernicus Marine login completed."
    )

    authentication_success = True

except Exception as e:

    authentication_success = False

    print(
        "\n✗ Copernicus Marine login failed."
    )

    print(
        "Error:"
    )

    print(
        e
    )

    raise RuntimeError(
        "Copernicus login failed. "
        "Do not continue to the data-download stage."
    )


# ======================================================================
# 8. DESCRIBE FUNCTION SIGNATURE
# ======================================================================

print("\n" + "=" * 70)
print("DESCRIBE FUNCTION")
print("=" * 70)

print(
    inspect.signature(
        copernicusmarine.describe
    )
)

print(
    "\n✓ describe() will now be called WITHOUT "
    "username/password arguments."
)


# ======================================================================
# 9. DATASET METADATA VERIFICATION
# ======================================================================

print("\n" + "=" * 70)
print("EXACT DATASET METADATA VERIFICATION")
print("=" * 70)

metadata_results = {}


for name, info in DATASETS.items():

    dataset_id = info["dataset_id"]

    print("\n" + "-" * 70)
    print(name)
    print("-" * 70)

    print(
        "Dataset ID:"
    )

    print(
        dataset_id
    )

    try:

        metadata = copernicusmarine.describe(
            dataset_id=dataset_id,
            disable_progress_bar=True
        )

        if hasattr(
            metadata,
            "model_dump"
        ):

            metadata_dict = (
                metadata.model_dump()
            )

        elif hasattr(
            metadata,
            "dict"
        ):

            metadata_dict = (
                metadata.dict()
            )

        elif isinstance(
            metadata,
            dict
        ):

            metadata_dict = metadata

        else:

            metadata_dict = {
                "representation":
                    repr(metadata)
            }

        metadata_results[name] = (
            metadata_dict
        )

        print(
            "✓ Metadata retrieved."
        )

    except Exception as e:

        metadata_results[name] = {
            "error": str(e)
        }

        print(
            "✗ Metadata retrieval failed."
        )

        print(
            e
        )


# ======================================================================
# 10. VARIABLE METADATA SEARCH
# ======================================================================

print("\n" + "=" * 70)
print("VARIABLE METADATA SEARCH")
print("=" * 70)


def find_variable_objects(
    obj,
    results=None
):

    if results is None:
        results = []

    if isinstance(
        obj,
        dict
    ):

        variable_keys = {
            "short_name",
            "standard_name",
            "long_name",
            "units"
        }

        if any(
            key in obj
            for key in variable_keys
        ):

            results.append(
                obj
            )

        for value in obj.values():

            find_variable_objects(
                value,
                results
            )

    elif isinstance(
        obj,
        list
    ):

        for value in obj:

            find_variable_objects(
                value,
                results
            )

    return results


variable_results = {}


for name, info in DATASETS.items():

    print("\n" + "-" * 70)
    print(name)
    print("-" * 70)

    records = find_variable_objects(
        metadata_results.get(
            name,
            {}
        )
    )

    variable_results[name] = []

    seen = set()

    for record in records:

        short_name = record.get(
            "short_name"
        )

        standard_name = record.get(
            "standard_name"
        )

        long_name = record.get(
            "long_name"
        )

        units = record.get(
            "units"
        )

        identity = (
            str(short_name),
            str(standard_name),
            str(long_name),
            str(units)
        )

        if identity in seen:
            continue

        seen.add(identity)

        cleaned = {
            "short_name":
                short_name,
            "standard_name":
                standard_name,
            "long_name":
                long_name,
            "units":
                units
        }

        variable_results[name].append(
            cleaned
        )

        print(
            "Short name    :",
            short_name
        )

        print(
            "Standard name :",
            standard_name
        )

        print(
            "Long name     :",
            long_name
        )

        print(
            "Units         :",
            units
        )

        print()


# ======================================================================
# 11. DEPTH / COORDINATE SEARCH
# ======================================================================

print("\n" + "=" * 70)
print("DEPTH / COORDINATE METADATA SEARCH")
print("=" * 70)


def recursive_key_search(
    obj,
    target_keys,
    results=None,
    path="root"
):

    if results is None:
        results = []

    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            key_lower = str(
                key
            ).lower()

            if key_lower in target_keys:

                results.append({
                    "path": path,
                    "key": str(key),
                    "value": value
                })

            recursive_key_search(
                value,
                target_keys,
                results,
                path + "." + str(key)
            )

    elif isinstance(
        obj,
        list
    ):

        for index, value in enumerate(
            obj
        ):

            recursive_key_search(
                value,
                target_keys,
                results,
                path + f"[{index}]"
            )

    return results


depth_results = {}


for name, metadata in (
    metadata_results.items()
):

    matches = recursive_key_search(
        metadata,
        {
            "depth",
            "depths",
            "depth_values",
            "depth_coordinates"
        }
    )

    depth_results[name] = matches

    print("\n" + name)
    print("-" * 70)

    if not matches:

        print(
            "No explicit depth coordinate "
            "found in catalogue metadata."
        )

    else:

        for match in matches[:10]:

            print(
                "Path :",
                match["path"]
            )

            print(
                "Key  :",
                match["key"]
            )

            print(
                "Value:",
                str(
                    match["value"]
                )[:500]
            )


# ======================================================================
# 12. DRY-RUN SUBSET VERIFICATION
# ======================================================================

print("\n" + "=" * 70)
print("DRY-RUN SUBSET VERIFICATION")
print("=" * 70)

print(
    "Validation period:"
)

print(
    f"{VALIDATION_START} → {VALIDATION_END}"
)

print(
    "\nStudy region:"
)

print(
    f"Latitude : {MIN_LATITUDE} → {MAX_LATITUDE}"
)

print(
    f"Longitude: {MIN_LONGITUDE} → {MAX_LONGITUDE}"
)

print(
    "\nNo scientific data will be downloaded."
)

print(
    "dry_run=True"
)


subset_results = {}


for name, info in DATASETS.items():

    dataset_id = info["dataset_id"]

    print("\n" + "-" * 70)
    print(
        f"DRY RUN — {name}"
    )
    print("-" * 70)

    try:

        result = copernicusmarine.subset(

            dataset_id=dataset_id,

            start_datetime=VALIDATION_START,
            end_datetime=VALIDATION_END,

            minimum_longitude=MIN_LONGITUDE,
            maximum_longitude=MAX_LONGITUDE,

            minimum_latitude=MIN_LATITUDE,
            maximum_latitude=MAX_LATITUDE,

            output_directory=str(
                EXTERNAL_VALIDATION_DIR
            ),

            dry_run=True,

            disable_progress_bar=True
        )

        subset_results[name] = {
            "success": True,
            "result": str(result)
        }

        print(
            "✓ Dry-run accepted."
        )

    except Exception as e:

        subset_results[name] = {
            "success": False,
            "error": str(e)
        }

        print(
            "✗ Dry-run failed."
        )

        print(
            e
        )


# ======================================================================
# 13. CHECK WHETHER FILES WERE ACTUALLY CREATED
# ======================================================================

print("\n" + "=" * 70)
print("DOWNLOAD SAFETY CHECK")
print("=" * 70)

unexpected_files = []

for path in EXTERNAL_VALIDATION_DIR.iterdir():

    if path.name.endswith(
        ".nc"
    ) or path.name.endswith(
        ".zarr"
    ):

        unexpected_files.append(
            str(path)
        )

print(
    "Scientific data files detected:"
)

print(
    len(unexpected_files)
)

if unexpected_files:

    print(
        "⚠ Unexpected data files found:"
    )

    for file in unexpected_files:
        print(
            file
        )

else:

    print(
        "✓ No scientific validation data "
        "were downloaded."
    )


# ======================================================================
# 14. SAVE VERIFICATION REPORT
# ======================================================================

verification_output = {

    "step":
        "31.9P-16T-23-C-FIX",

    "validation_period": {
        "start":
            VALIDATION_START,
        "end":
            VALIDATION_END
    },

    "study_region": {
        "minimum_latitude":
            MIN_LATITUDE,
        "maximum_latitude":
            MAX_LATITUDE,
        "minimum_longitude":
            MIN_LONGITUDE,
        "maximum_longitude":
            MAX_LONGITUDE
    },

    "datasets":
        DATASETS,

    "authentication_success":
        authentication_success,

    "metadata_results":
        metadata_results,

    "variable_results":
        variable_results,

    "depth_results":
        depth_results,

    "subset_results":
        subset_results,

    "scientific_data_downloaded":
        False,

    "final_model_modified":
        False,

    "shap_modified":
        False,

    "three_d_visualization_modified":
        False
}


with open(
    OUTPUT_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        verification_output,
        file,
        indent=2,
        default=str
    )


print("\n" + "=" * 70)
print("VERIFICATION REPORT SAVED")
print("=" * 70)

print(
    OUTPUT_FILE
)


# ======================================================================
# 15. FINAL SAFETY STATUS
# ======================================================================

print("\n" + "=" * 70)
print("FINAL 31.9P-16T-23-C-FIX STATUS")
print("=" * 70)

print(
    "Copernicus login successful :",
    authentication_success
)

print(
    "Dataset metadata checked    : TRUE"
)

print(
    "Variable metadata checked   : TRUE"
)

print(
    "Depth metadata checked      : TRUE"
)

print(
    "2026 period requested       : TRUE"
)

print(
    "Study region requested      : TRUE"
)

print(
    "Dry-run performed           : TRUE"
)

print(
    "Scientific data downloaded  : FALSE"
)

print(
    "XGBoost retrained           : FALSE"
)

print(
    "XGBoost modified            : FALSE"
)

print(
    "SHAP recalculated           : FALSE"
)

print(
    "3D visualization modified   : FALSE"
)

print("\n" + "=" * 70)
print("STEP 31.9P-16T-23-C-FIX COMPLETE")
print("=" * 70)

print(
    "\nNEXT:"
)

print(
    "Send the COMPLETE output."
)

print(
    "Do NOT download the 2026 validation "
    "dataset yet."
)
```

```python
# ======================================================================
# STEP 31.9P-16T-24-D
# MULTI-DAY COPERNICUS CHL COVERAGE TEST
# ======================================================================

print("=" * 70)
print("STEP 31.9P-16T-24-D")
print("MULTI-DAY COPERNICUS CHL COVERAGE TEST")
print("=" * 70)


# ======================================================================
# 1. IMPORTS
# ======================================================================

from pathlib import Path
import json
import shutil
import gc

import numpy as np
import pandas as pd
import xarray as xr
import copernicusmarine


# ======================================================================
# 2. CONFIGURATION
# ======================================================================

DATASET_ID = (
    "cmems_obs-oc_glo_bgc-plankton_my_l3-multi-4km_P1D"
)

DATASET_VERSION = "202611"


# Seven representative dates
TEST_DATES = [
    "2026-01-01",
    "2026-01-15",
    "2026-02-01",
    "2026-02-15",
    "2026-03-01",
    "2026-03-15",
    "2026-04-01"
]


# Sri Lankan Western + Southern study region
LAT_MIN = 5.8333335
LAT_MAX = 6.9166665

LON_MIN = 79.333336
LON_MAX = 80.0


# Project directories
PROJECT_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

EXTERNAL_DIR = (
    PROJECT_DIR
    / "06_models/02_XGBoost/"
      "hyperparameter_optimization/"
      "final_tuned_candidate/"
      "external_validation"
)

COVERAGE_DIR = (
    EXTERNAL_DIR
    / "copernicus_chl_2026_coverage_test"
)

TEMP_DIR = (
    COVERAGE_DIR
    / "temporary_original_files"
)

SUBSET_DIR = (
    COVERAGE_DIR
    / "regional_chl_subsets"
)

COVERAGE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TEMP_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SUBSET_DIR.mkdir(
    parents=True,
    exist_ok=True
)


REPORT_FILE = (
    EXTERNAL_DIR
    / "copernicus_chl_2026_multiday_coverage_verification.json"
)

SUMMARY_FILE = (
    COVERAGE_DIR
    / "copernicus_chl_2026_multiday_coverage_summary.csv"
)


# ======================================================================
# 3. INITIAL INFORMATION
# ======================================================================

print("\n" + "=" * 70)
print("TEST CONFIGURATION")
print("=" * 70)

print(
    "Dataset:",
    DATASET_ID
)

print(
    "Dataset version:",
    DATASET_VERSION
)

print(
    "\nTest dates:"
)

for date in TEST_DATES:

    print(
        " -",
        date
    )

print(
    "\nNumber of test dates:",
    len(TEST_DATES)
)

print(
    "\nStudy region:"
)

print(
    "Latitude :",
    LAT_MIN,
    "→",
    LAT_MAX
)

print(
    "Longitude:",
    LON_MIN,
    "→",
    LON_MAX
)

print(
    "\nIMPORTANT:"
)

print(
    "Only 7 original files will be retrieved."
)

print(
    "No full Jan-Apr download will be performed."
)

print(
    "No XGBoost modification will occur."
)

print(
    "No SHAP calculation will occur."
)

print(
    "No 3D visualization modification will occur."
)


# ======================================================================
# 4. RESULT STORAGE
# ======================================================================

results = []

total_original_bytes = 0
total_subset_bytes = 0


# ======================================================================
# 5. PROCESS EACH TEST DATE
# ======================================================================

for day_number, target_date in enumerate(
    TEST_DATES,
    start=1
):

    print("\n")
    print("=" * 70)
    print(
        f"TEST DATE {day_number}/{len(TEST_DATES)}"
    )
    print(
        target_date
    )
    print("=" * 70)


    # --------------------------------------------------------------
    # Date formatting
    # --------------------------------------------------------------

    date_compact = (
        pd.Timestamp(target_date)
        .strftime("%Y%m%d")
    )


    # --------------------------------------------------------------
    # Temporary directory for this date
    # --------------------------------------------------------------

    date_temp_dir = (
        TEMP_DIR
        / date_compact
    )

    date_temp_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------------
    # Exact file filter
    # --------------------------------------------------------------

    file_filter = (
        f"*{date_compact}_"
        "cmems_obs-oc_glo_bgc-plankton_my_l3-multi-4km_P1D.nc"
    )


    print(
        "\nFile filter:"
    )

    print(
        file_filter
    )


    # ==============================================================
    # 5A. DRY RUN
    # ==============================================================

    print("\n" + "-" * 70)
    print("DRY RUN")
    print("-" * 70)

    try:

        dry_run = copernicusmarine.get(

            dataset_id=DATASET_ID,

            dataset_version=DATASET_VERSION,

            filter=file_filter,

            output_directory=str(
                date_temp_dir
            ),

            dry_run=True,

            disable_progress_bar=False
        )

        print(
            "✓ Dry run completed."
        )

    except Exception as e:

        print(
            "✗ Dry run failed:"
        )

        print(
            repr(e)
        )

        raise


    # ==============================================================
    # 5B. DOWNLOAD ONE ORIGINAL FILE
    # ==============================================================

    print("\n" + "-" * 70)
    print("DOWNLOADING ONE ORIGINAL FILE")
    print("-" * 70)

    try:

        copernicusmarine.get(

            dataset_id=DATASET_ID,

            dataset_version=DATASET_VERSION,

            filter=file_filter,

            output_directory=str(
                date_temp_dir
            ),

            skip_existing=True,

            disable_progress_bar=False
        )

        print(
            "✓ Original file downloaded."
        )

    except Exception as e:

        print(
            "✗ Download failed:"
        )

        print(
            repr(e)
        )

        raise


    # ==============================================================
    # 5C. FIND NETCDF FILE
    # ==============================================================

    nc_files = sorted(
        date_temp_dir.rglob(
            "*.nc"
        )
    )


    print(
        "\nNetCDF files found:",
        len(nc_files)
    )


    if len(nc_files) != 1:

        raise RuntimeError(
            f"Expected exactly 1 NetCDF file "
            f"for {target_date}, "
            f"but found {len(nc_files)}."
        )


    original_file = nc_files[0]


    original_size_bytes = (
        original_file.stat().st_size
    )

    original_size_mb = (
        original_size_bytes
        / (1024 ** 2)
    )


    print(
        "Original file:"
    )

    print(
        original_file
    )

    print(
        f"Original size: "
        f"{original_size_mb:.2f} MB"
    )


    # ==============================================================
    # 5D. OPEN ORIGINAL FILE
    # ==============================================================

    print("\n" + "-" * 70)
    print("OPENING ORIGINAL NETCDF")
    print("-" * 70)

    ds = xr.open_dataset(
        original_file
    )


    # ==============================================================
    # 5E. VARIABLE CHECK
    # ==============================================================

    if "CHL" not in ds.data_vars:

        ds.close()

        raise RuntimeError(
            f"CHL variable not found for {target_date}."
        )


    chl = ds["CHL"]


    # ==============================================================
    # 5F. COORDINATE CHECK
    # ==============================================================

    LAT_NAME = None
    LON_NAME = None


    for candidate in [
        "latitude",
        "lat",
        "nav_lat",
        "y"
    ]:

        if candidate in ds.coords:

            LAT_NAME = candidate

            break


    for candidate in [
        "longitude",
        "lon",
        "nav_lon",
        "x"
    ]:

        if candidate in ds.coords:

            LON_NAME = candidate

            break


    if LAT_NAME is None:

        ds.close()

        raise RuntimeError(
            f"Latitude coordinate not found "
            f"for {target_date}."
        )


    if LON_NAME is None:

        ds.close()

        raise RuntimeError(
            f"Longitude coordinate not found "
            f"for {target_date}."
        )


    # ==============================================================
    # 5G. LOCAL SPATIAL SUBSET
    # ==============================================================

    lat_values = ds[
        LAT_NAME
    ].values

    lon_values = ds[
        LON_NAME
    ].values


    if lat_values[0] < lat_values[-1]:

        lat_slice = slice(
            LAT_MIN,
            LAT_MAX
        )

    else:

        lat_slice = slice(
            LAT_MAX,
            LAT_MIN
        )


    if lon_values[0] < lon_values[-1]:

        lon_slice = slice(
            LON_MIN,
            LON_MAX
        )

    else:

        lon_slice = slice(
            LON_MAX,
            LON_MIN
        )


    regional = ds.sel(
        {
            LAT_NAME: lat_slice,
            LON_NAME: lon_slice
        }
    )


    regional_chl = regional[
        "CHL"
    ]


    # ==============================================================
    # 5H. CHL VALUES
    # ==============================================================

    chl_values = (
        regional_chl
        .values
    )


    total_cells = (
        chl_values.size
    )


    finite_mask = np.isfinite(
        chl_values
    )


    valid_values = (
        chl_values[
            finite_mask
        ]
    )


    valid_cells = (
        valid_values.size
    )


    missing_cells = (
        total_cells
        - valid_cells
    )


    if total_cells > 0:

        valid_percentage = (
            valid_cells
            / total_cells
        ) * 100.0

    else:

        valid_percentage = 0.0


    # ==============================================================
    # 5I. CHL STATISTICS
    # ==============================================================

    if valid_cells > 0:

        chl_min = float(
            np.min(
                valid_values
            )
        )

        chl_max = float(
            np.max(
                valid_values
            )
        )

        chl_mean = float(
            np.mean(
                valid_values
            )
        )

    else:

        chl_min = None
        chl_max = None
        chl_mean = None


    # ==============================================================
    # 5J. TIME CHECK
    # ==============================================================

    if "time" in regional.coords:

        time_values = (
            regional["time"]
            .values
        )

        time_text = [
            str(x)
            for x in time_values
        ]

    else:

        time_text = []


    # ==============================================================
    # 5K. SAVE ONLY CHL REGIONAL SUBSET
    # ==============================================================

    print("\n" + "-" * 70)
    print("SAVING REGIONAL CHL SUBSET")
    print("-" * 70)


    # Keep only the information required for
    # external CHL validation:
    #
    # time + latitude + longitude + CHL

    regional_chl_only = (
        regional[
            ["CHL"]
        ]
    )


    subset_file = (
        SUBSET_DIR
        / f"copernicus_chl_"
          f"{date_compact}_"
          f"sri_lanka.nc"
    )


    regional_chl_only.to_netcdf(
        subset_file
    )


    subset_size_bytes = (
        subset_file.stat().st_size
    )

    subset_size_mb = (
        subset_size_bytes
        / (1024 ** 2)
    )


    print(
        "✓ Regional CHL subset saved."
    )

    print(
        subset_file
    )

    print(
        f"Subset size: "
        f"{subset_size_mb:.4f} MB"
    )


    # ==============================================================
    # 5L. STORE RESULTS
    # ==============================================================

    result = {

        "date":
            target_date,

        "original_file_size_mb":
            round(
                original_size_mb,
                4
            ),

        "subset_file_size_mb":
            round(
                subset_size_mb,
                4
            ),

        "total_cells":
            int(
                total_cells
            ),

        "valid_cells":
            int(
                valid_cells
            ),

        "missing_cells":
            int(
                missing_cells
            ),

        "valid_percentage":
            round(
                valid_percentage,
                4
            ),

        "chl_min":
            chl_min,

        "chl_max":
            chl_max,

        "chl_mean":
            chl_mean,

        "subset_file":
            str(
                subset_file
            ),

        "time":
            time_text,

        "status":
            "SUCCESS"
    }


    results.append(
        result
    )


    total_original_bytes += (
        original_size_bytes
    )

    total_subset_bytes += (
        subset_size_bytes
    )


    # ==============================================================
    # 5M. PRINT DAILY RESULT
    # ==============================================================

    print("\n" + "-" * 70)
    print(
        f"RESULT — {target_date}"
    )
    print("-" * 70)

    print(
        "Total cells:",
        total_cells
    )

    print(
        "Valid CHL cells:",
        valid_cells
    )

    print(
        "Missing/non-finite:",
        missing_cells
    )

    print(
        f"Valid percentage: "
        f"{valid_percentage:.2f}%"
    )


    if valid_cells > 0:

        print(
            f"CHL minimum: "
            f"{chl_min:.6f}"
        )

        print(
            f"CHL maximum: "
            f"{chl_max:.6f}"
        )

        print(
            f"CHL mean: "
            f"{chl_mean:.6f}"
        )

    else:

        print(
            "No valid CHL observations."
        )


    # ==============================================================
    # 5N. CLOSE DATASET
    # ==============================================================

    ds.close()

    del regional
    del regional_chl
    del regional_chl_only
    del chl
    del ds

    gc.collect()


    # ==============================================================
    # 5O. DELETE LARGE ORIGINAL FILE
    # ==============================================================

    print("\n" + "-" * 70)
    print("REMOVING LARGE ORIGINAL FILE")
    print("-" * 70)


    try:

        shutil.rmtree(
            date_temp_dir
        )

        print(
            "✓ Large original-file directory removed."
        )

    except Exception as e:

        print(
            "⚠ Could not completely remove "
            "temporary original file:"
        )

        print(
            repr(e)
        )


# ======================================================================
# 6. CREATE SUMMARY DATAFRAME
# ======================================================================

print("\n" + "=" * 70)
print("MULTI-DAY COVERAGE SUMMARY")
print("=" * 70)


summary_df = pd.DataFrame(
    results
)


print(
    summary_df[
        [
            "date",
            "original_file_size_mb",
            "subset_file_size_mb",
            "total_cells",
            "valid_cells",
            "missing_cells",
            "valid_percentage",
            "chl_min",
            "chl_max",
            "chl_mean"
        ]
    ].to_string(
        index=False
    )
)


# ======================================================================
# 7. OVERALL COVERAGE STATISTICS
# ======================================================================

print("\n" + "=" * 70)
print("OVERALL COVERAGE ANALYSIS")
print("=" * 70)


mean_valid_percentage = (
    summary_df[
        "valid_percentage"
    ].mean()
)


median_valid_percentage = (
    summary_df[
        "valid_percentage"
    ].median()
)


days_with_valid_chl = (
    summary_df[
        "valid_cells"
    ] > 0
).sum()


days_with_no_valid_chl = (
    summary_df[
        "valid_cells"
    ] == 0
).sum()


print(
    f"Days tested: "
    f"{len(summary_df)}"
)

print(
    f"Days with valid CHL: "
    f"{days_with_valid_chl}"
)

print(
    f"Days with no valid CHL: "
    f"{days_with_no_valid_chl}"
)

print(
    f"Mean valid-cell percentage: "
    f"{mean_valid_percentage:.2f}%"
)

print(
    f"Median valid-cell percentage: "
    f"{median_valid_percentage:.2f}%"
)


# ======================================================================
# 8. ESTIMATE FULL JAN-APR STORAGE
# ======================================================================

print("\n" + "=" * 70)
print("JAN-APR STORAGE ESTIMATE")
print("=" * 70)


# Jan-Apr 2026 contains 120 calendar days.
FULL_PERIOD_DAYS = 120


mean_original_mb = (
    summary_df[
        "original_file_size_mb"
    ].mean()
)


mean_subset_mb = (
    summary_df[
        "subset_file_size_mb"
    ].mean()
)


estimated_original_gb = (
    mean_original_mb
    * FULL_PERIOD_DAYS
    / 1024
)


estimated_subset_mb = (
    mean_subset_mb
    * FULL_PERIOD_DAYS
)


estimated_subset_gb = (
    estimated_subset_mb
    / 1024
)


print(
    f"Mean original daily file: "
    f"{mean_original_mb:.2f} MB"
)

print(
    f"Estimated 120-day original storage: "
    f"{estimated_original_gb:.2f} GB"
)

print(
    f"Mean regional subset per day: "
    f"{mean_subset_mb:.4f} MB"
)

print(
    f"Estimated 120-day regional storage: "
    f"{estimated_subset_mb:.2f} MB"
)

print(
    f"Estimated 120-day regional storage: "
    f"{estimated_subset_gb:.4f} GB"
)


# ======================================================================
# 9. SAVE SUMMARY CSV
# ======================================================================

summary_df.to_csv(
    SUMMARY_FILE,
    index=False
)


print(
    "\n✓ Coverage summary saved:"
)

print(
    SUMMARY_FILE
)


# ======================================================================
# 10. DETERMINE PRELIMINARY COVERAGE STATUS
# ======================================================================

# IMPORTANT:
# This is NOT the final scientific validation decision.
# It only tells us whether the tested days contain any usable CHL.

if days_with_valid_chl == len(
    summary_df
):

    preliminary_status = (
        "VALID_CHL_PRESENT_ON_ALL_TEST_DAYS"
    )

elif days_with_valid_chl > 0:

    preliminary_status = (
        "PARTIAL_CHL_COVERAGE"
    )

else:

    preliminary_status = (
        "NO_VALID_CHL_ON_TEST_DAYS"
    )


print("\n" + "=" * 70)
print("PRELIMINARY COVERAGE STATUS")
print("=" * 70)

print(
    preliminary_status
)


# ======================================================================
# 11. SAVE JSON REPORT
# ======================================================================

report = {

    "step":
        "31.9P-16T-24-D",

    "dataset_id":
        DATASET_ID,

    "dataset_version":
        DATASET_VERSION,

    "test_dates":
        TEST_DATES,

    "study_region": {

        "latitude_min":
            LAT_MIN,

        "latitude_max":
            LAT_MAX,

        "longitude_min":
            LON_MIN,

        "longitude_max":
            LON_MAX
    },

    "results":
        results,

    "overall": {

        "days_tested":
            int(
                len(summary_df)
            ),

        "days_with_valid_chl":
            int(
                days_with_valid_chl
            ),

        "days_with_no_valid_chl":
            int(
                days_with_no_valid_chl
            ),

        "mean_valid_percentage":
            float(
                mean_valid_percentage
            ),

        "median_valid_percentage":
            float(
                median_valid_percentage
            )
    },

    "storage_estimate": {

        "mean_original_file_mb":
            float(
                mean_original_mb
            ),

        "estimated_120_day_original_gb":
            float(
                estimated_original_gb
            ),

        "mean_regional_subset_mb":
            float(
                mean_subset_mb
            ),

        "estimated_120_day_regional_mb":
            float(
                estimated_subset_mb
            ),

        "estimated_120_day_regional_gb":
            float(
                estimated_subset_gb
            )
    },

    "preliminary_status":
        preliminary_status,

    "full_jan_apr_download":
        False,

    "all_120_original_files_retained":
        False,

    "xgboost_modified":
        False,

    "xgboost_retrained":
        False,

    "shap_modified":
        False,

    "three_d_visualization_modified":
        False
}


with open(
    REPORT_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        report,
        f,
        indent=2
    )


# ======================================================================
# 12. FINAL SAFETY CHECK
# ======================================================================

print("\n" + "=" * 70)
print("FINAL 31.9P-16T-24-D SAFETY CHECK")
print("=" * 70)

print(
    "7 representative dates tested      : TRUE"
)

print(
    "Sri Lankan spatial subsets created : TRUE"
)

print(
    "Valid CHL coverage calculated      : TRUE"
)

print(
    "Storage estimate calculated        : TRUE"
)

print(
    "Full Jan-Apr download              : FALSE"
)

print(
    "All 120 original files retained    : FALSE"
)

print(
    "Final XGBoost modified             : FALSE"
)

print(
    "Final XGBoost retrained            : FALSE"
)

print(
    "SHAP modified                      : FALSE"
)

print(
    "3D visualization modified          : FALSE"
)


# ======================================================================
# 13. COMPLETE
# ======================================================================

print("\n" + "=" * 70)
print("STEP 31.9P-16T-24-D COMPLETE")
print("=" * 70)

print(
    "\nCoverage summary:"
)

print(
    SUMMARY_FILE
)

print(
    "\nVerification report:"
)

print(
    REPORT_FILE
)

print(
    "\nIMPORTANT:"
)

print(
    "DO NOT download the remaining Jan-Apr files yet."
)

print(
    "Send me the COMPLETE output first."
)
```

```python
# ======================================================================
# STEP 31.9P-16T-24-F
# FULL JAN-APR 2026 REGIONAL COPERNICUS CHL PREPARATION
# ======================================================================

print("=" * 70)
print("STEP 31.9P-16T-24-F")
print("FULL JAN-APR 2026 REGIONAL COPERNICUS CHL PREPARATION")
print("=" * 70)


# ======================================================================
# 1. IMPORTS
# ======================================================================

from pathlib import Path
import json
import shutil
import gc
import time
import os

import numpy as np
import pandas as pd
import xarray as xr
import copernicusmarine


# ======================================================================
# 2. CONFIGURATION
# ======================================================================

DATASET_ID = (
    "cmems_obs-oc_glo_bgc-plankton_my_l3-multi-4km_P1D"
)

DATASET_VERSION = "202611"


# Full external-validation period
START_DATE = pd.Timestamp(
    "2026-01-01"
)

END_DATE = pd.Timestamp(
    "2026-04-30"
)


# Western + Southern Sri Lanka
LAT_MIN = 5.8333335
LAT_MAX = 6.9166665

LON_MIN = 79.333336
LON_MAX = 80.0


# ======================================================================
# 3. PROJECT DIRECTORIES
# ======================================================================

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

MODEL_DIR = (
    PROJECT_DIR
    / "06_models/02_XGBoost/"
      "hyperparameter_optimization/"
      "final_tuned_candidate"
)

EXTERNAL_DIR = (
    MODEL_DIR
    / "external_validation"
)

FINAL_DATA_DIR = (
    EXTERNAL_DIR
    / "copernicus_chl_2026_jan_apr"
)

REGIONAL_DIR = (
    FINAL_DATA_DIR
    / "regional_daily_chl"
)

TEMP_DIR = (
    FINAL_DATA_DIR
    / "temporary_original_files"
)


EXTERNAL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FINAL_DATA_DIR.mkdir(
    parents=True,
    exist_ok=True
)

REGIONAL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TEMP_DIR.mkdir(
    parents=True,
    exist_ok=True
)


SUMMARY_CSV = (
    FINAL_DATA_DIR
    / "copernicus_chl_2026_jan_apr_daily_summary.csv"
)

REPORT_JSON = (
    EXTERNAL_DIR
    / "copernicus_chl_2026_jan_apr_preparation_report.json"
)


# ======================================================================
# 4. DATE LIST
# ======================================================================

DATES = pd.date_range(
    START_DATE,
    END_DATE,
    freq="D"
)

EXPECTED_DAYS = len(
    DATES
)


# ======================================================================
# 5. HEADER
# ======================================================================

print("\n" + "=" * 70)
print("CONFIGURATION")
print("=" * 70)

print(
    "Dataset:",
    DATASET_ID
)

print(
    "Dataset version:",
    DATASET_VERSION
)

print(
    "Validation start:",
    START_DATE.strftime("%Y-%m-%d")
)

print(
    "Validation end:",
    END_DATE.strftime("%Y-%m-%d")
)

print(
    "Expected calendar days:",
    EXPECTED_DAYS
)

print(
    "\nStudy region:"
)

print(
    "Latitude :",
    LAT_MIN,
    "→",
    LAT_MAX
)

print(
    "Longitude:",
    LON_MIN,
    "→",
    LON_MAX
)


# ======================================================================
# 6. SAFETY / MODEL ARTIFACT CHECK
# ======================================================================

print("\n" + "=" * 70)
print("FINAL RESEARCH ARTIFACT CHECK")
print("=" * 70)


# Search for the model using the known project tree.
# This is only a presence check.

possible_model_files = list(
    MODEL_DIR.rglob(
        "*.json"
    )
)


possible_feature_files = list(
    MODEL_DIR.rglob(
        "*feature*"
    )
)


print(
    "Model directory exists:",
    MODEL_DIR.exists()
)

print(
    "External validation directory exists:",
    EXTERNAL_DIR.exists()
)

print(
    "Possible model/config JSON files:",
    len(possible_model_files)
)

print(
    "Possible feature-list files:",
    len(possible_feature_files)
)


print(
    "\n✓ This step will not load or modify the XGBoost model."
)

print(
    "✓ This step will not retrain the model."
)

print(
    "✓ This step will not regenerate predictions."
)

print(
    "✓ This step will not modify SHAP outputs."
)

print(
    "✓ This step will not modify the 3D visualization."
)


# ======================================================================
# 7. RESULT STORAGE
# ======================================================================

results = []

successful_days = 0

failed_days = 0

days_with_valid_chl = 0

days_with_no_valid_chl = 0

total_original_bytes = 0

total_regional_bytes = 0


# ======================================================================
# 8. PROCESS EACH DAY
# ======================================================================

for day_index, current_date in enumerate(
    DATES,
    start=1
):

    date_text = (
        current_date.strftime(
            "%Y-%m-%d"
        )
    )

    date_compact = (
        current_date.strftime(
            "%Y%m%d"
        )
    )


    print("\n")
    print("=" * 70)
    print(
        f"DAY {day_index}/{EXPECTED_DAYS} — {date_text}"
    )
    print("=" * 70)


    # --------------------------------------------------------------
    # Temporary directory for this day
    # --------------------------------------------------------------

    day_temp_dir = (
        TEMP_DIR
        / date_compact
    )

    day_temp_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------------
    # File filter
    # --------------------------------------------------------------

    file_filter = (
        f"*{date_compact}_"
        "cmems_obs-oc_glo_bgc-plankton_my_l3-multi-4km_P1D.nc"
    )


    print(
        "Remote file filter:"
    )

    print(
        file_filter
    )


    # --------------------------------------------------------------
    # Result record
    # --------------------------------------------------------------

    result = {

        "date":
            date_text,

        "status":
            "NOT_PROCESSED",

        "original_file_size_mb":
            None,

        "regional_file_size_mb":
            None,

        "total_cells":
            0,

        "valid_cells":
            0,

        "missing_cells":
            0,

        "valid_percentage":
            0.0,

        "chl_min":
            None,

        "chl_max":
            None,

        "chl_mean":
            None,

        "regional_file":
            None,

        "error":
            None
    }


    ds = None


    try:

        # ==========================================================
        # 8A. DOWNLOAD ONE ORIGINAL FILE
        # ==========================================================

        print(
            "\nDownloading one temporary original file..."
        )

        download_start = time.time()


        copernicusmarine.get(

            dataset_id=DATASET_ID,

            dataset_version=DATASET_VERSION,

            filter=file_filter,

            output_directory=str(
                day_temp_dir
            ),

            skip_existing=True,

            disable_progress_bar=False
        )


        download_elapsed = (
            time.time()
            - download_start
        )


        print(
            f"✓ Download completed "
            f"in {download_elapsed:.1f} seconds."
        )


        # ==========================================================
        # 8B. FIND EXACT NETCDF FILE
        # ==========================================================

        nc_files = sorted(
            day_temp_dir.rglob(
                "*.nc"
            )
        )


        if len(nc_files) != 1:

            raise RuntimeError(
                f"Expected exactly 1 NetCDF file "
                f"for {date_text}, "
                f"but found {len(nc_files)}."
            )


        original_file = nc_files[0]


        original_size_bytes = (
            original_file.stat().st_size
        )

        original_size_mb = (
            original_size_bytes
            / (1024 ** 2)
        )


        result[
            "original_file_size_mb"
        ] = round(
            original_size_mb,
            4
        )


        total_original_bytes += (
            original_size_bytes
        )


        print(
            f"Original file size: "
            f"{original_size_mb:.2f} MB"
        )


        # ==========================================================
        # 8C. OPEN ORIGINAL FILE
        # ==========================================================

        ds = xr.open_dataset(
            original_file
        )


        # ==========================================================
        # 8D. VERIFY CHL
        # ==========================================================

        if "CHL" not in ds.data_vars:

            raise RuntimeError(
                f"CHL variable not found "
                f"for {date_text}."
            )


        chl = ds["CHL"]


        # ==========================================================
        # 8E. FIND COORDINATES
        # ==========================================================

        LAT_NAME = None

        for candidate in [
            "latitude",
            "lat",
            "nav_lat",
            "y"
        ]:

            if candidate in ds.coords:

                LAT_NAME = candidate

                break


        LON_NAME = None

        for candidate in [
            "longitude",
            "lon",
            "nav_lon",
            "x"
        ]:

            if candidate in ds.coords:

                LON_NAME = candidate

                break


        if LAT_NAME is None:

            raise RuntimeError(
                f"Latitude coordinate not found "
                f"for {date_text}."
            )


        if LON_NAME is None:

            raise RuntimeError(
                f"Longitude coordinate not found "
                f"for {date_text}."
            )


        # ==========================================================
        # 8F. CREATE LOCAL SPATIAL SUBSET
        # ==========================================================

        lat_values = ds[
            LAT_NAME
        ].values

        lon_values = ds[
            LON_NAME
        ].values


        if lat_values[0] < lat_values[-1]:

            lat_slice = slice(
                LAT_MIN,
                LAT_MAX
            )

        else:

            lat_slice = slice(
                LAT_MAX,
                LAT_MIN
            )


        if lon_values[0] < lon_values[-1]:

            lon_slice = slice(
                LON_MIN,
                LON_MAX
            )

        else:

            lon_slice = slice(
                LON_MAX,
                LON_MIN
            )


        regional = ds.sel(
            {
                LAT_NAME: lat_slice,
                LON_NAME: lon_slice
            }
        )


        # ==========================================================
        # 8G. KEEP ONLY CHL
        # ==========================================================

        regional_chl = regional[
            ["CHL"]
        ]


        # ==========================================================
        # 8H. LOAD ONLY SMALL REGIONAL ARRAY
        # ==========================================================

        regional_chl = (
            regional_chl.load()
        )


        chl_values = (
            regional_chl[
                "CHL"
            ].values
        )


        total_cells = (
            chl_values.size
        )


        finite_mask = np.isfinite(
            chl_values
        )


        valid_values = (
            chl_values[
                finite_mask
            ]
        )


        valid_cells = (
            valid_values.size
        )


        missing_cells = (
            total_cells
            - valid_cells
        )


        if total_cells > 0:

            valid_percentage = (
                valid_cells
                / total_cells
            ) * 100.0

        else:

            valid_percentage = 0.0


        result[
            "total_cells"
        ] = int(
            total_cells
        )

        result[
            "valid_cells"
        ] = int(
            valid_cells
        )

        result[
            "missing_cells"
        ] = int(
            missing_cells
        )

        result[
            "valid_percentage"
        ] = round(
            valid_percentage,
            4
        )


        # ==========================================================
        # 8I. VALID CHL STATISTICS
        # ==========================================================

        if valid_cells > 0:

            chl_min = float(
                np.min(
                    valid_values
                )
            )

            chl_max = float(
                np.max(
                    valid_values
                )
            )

            chl_mean = float(
                np.mean(
                    valid_values
                )
            )


            result[
                "chl_min"
            ] = chl_min

            result[
                "chl_max"
            ] = chl_max

            result[
                "chl_mean"
            ] = chl_mean


            days_with_valid_chl += 1


        else:

            # IMPORTANT:
            # No valid CHL is represented as missing data.
            # It is NOT converted to zero.

            days_with_no_valid_chl += 1


        # ==========================================================
        # 8J. SAVE REGIONAL FILE
        # ==========================================================

        regional_file = (
            REGIONAL_DIR
            / f"copernicus_chl_"
              f"{date_compact}_"
              f"sri_lanka.nc"
        )


        regional_chl.to_netcdf(
            regional_file
        )


        regional_size_bytes = (
            regional_file.stat().st_size
        )

        regional_size_mb = (
            regional_size_bytes
            / (1024 ** 2)
        )


        result[
            "regional_file_size_mb"
        ] = round(
            regional_size_mb,
            6
        )


        result[
            "regional_file"
        ] = str(
            regional_file
        )


        total_regional_bytes += (
            regional_size_bytes
        )


        # ==========================================================
        # 8K. CLOSE DATASET
        # ==========================================================

        try:

            ds.close()

        except Exception:

            pass


        ds = None


        del regional
        del regional_chl
        del chl
        del chl_values
        del valid_values

        gc.collect()


        # ==========================================================
        # 8L. DELETE TEMPORARY GLOBAL FILE
        # ==========================================================

        print(
            "\nDeleting temporary global file..."
        )


        shutil.rmtree(
            day_temp_dir
        )


        print(
            "✓ Temporary global file removed."
        )


        # ==========================================================
        # 8M. MARK SUCCESS
        # ==========================================================

        result[
            "status"
        ] = "SUCCESS"


        successful_days += 1


        print(
            "\n✓ Day completed successfully."
        )


        print(
            f"Total cells: "
            f"{total_cells}"
        )

        print(
            f"Valid CHL cells: "
            f"{valid_cells}"
        )

        print(
            f"Missing/non-finite: "
            f"{missing_cells}"
        )

        print(
            f"Valid percentage: "
            f"{valid_percentage:.2f}%"
        )


        if valid_cells > 0:

            print(
                f"CHL min: "
                f"{chl_min:.6f}"
            )

            print(
                f"CHL max: "
                f"{chl_max:.6f}"
            )

            print(
                f"CHL mean: "
                f"{chl_mean:.6f}"
            )

        else:

            print(
                "⚠ No valid CHL observations "
                "for this date."
            )

            print(
                "Date retained as missing."
            )


    except Exception as e:

        failed_days += 1


        result[
            "status"
        ] = "FAILED"


        result[
            "error"
        ] = repr(e)


        print(
            "\n✗ ERROR PROCESSING DATE:"
        )

        print(
            date_text
        )

        print(
            repr(e)
        )


        # Close open dataset if necessary

        try:

            if ds is not None:

                ds.close()

        except Exception:

            pass


        ds = None


        # Remove temporary global file
        # before continuing to the next date.

        try:

            if day_temp_dir.exists():

                shutil.rmtree(
                    day_temp_dir
                )

        except Exception as cleanup_error:

            print(
                "⚠ Temporary cleanup failed:"
            )

            print(
                repr(
                    cleanup_error
                )
            )


        gc.collect()


        print(
            "\nContinuing to next date..."
        )


    # ==============================================================
    # 8N. APPEND RESULT
    # ==============================================================

    results.append(
        result
    )


    # ==============================================================
    # 8O. SAVE PROGRESS AFTER EVERY DATE
    # ==============================================================

    progress_df = pd.DataFrame(
        results
    )


    progress_df.to_csv(
        SUMMARY_CSV,
        index=False
    )


    # ==============================================================
    # 8P. PROGRESS SUMMARY
    # ==============================================================

    print("\n" + "-" * 70)
    print("RUNNING PROGRESS")
    print("-" * 70)

    print(
        f"Processed: "
        f"{day_index}/{EXPECTED_DAYS}"
    )

    print(
        f"Successful: "
        f"{successful_days}"
    )

    print(
        f"Failed: "
        f"{failed_days}"
    )

    print(
        f"Days with valid CHL: "
        f"{days_with_valid_chl}"
    )

    print(
        f"Days with no valid CHL: "
        f"{days_with_no_valid_chl}"
    )


# ======================================================================
# 9. FINAL SUMMARY
# ======================================================================

summary_df = pd.DataFrame(
    results
)


print("\n" + "=" * 70)
print("FINAL JAN-APR 2026 PREPARATION SUMMARY")
print("=" * 70)


print(
    "Expected days:",
    EXPECTED_DAYS
)

print(
    "Processed days:",
    len(summary_df)
)

print(
    "Successful days:",
    successful_days
)

print(
    "Failed days:",
    failed_days
)

print(
    "Days with valid CHL:",
    days_with_valid_chl
)

print(
    "Days with no valid CHL:",
    days_with_no_valid_chl
)


# ======================================================================
# 10. COVERAGE STATISTICS
# ======================================================================

if len(summary_df) > 0:

    valid_percentage_mean = (
        summary_df[
            "valid_percentage"
        ].mean()
    )

    valid_percentage_median = (
        summary_df[
            "valid_percentage"
        ].median()
    )

else:

    valid_percentage_mean = 0.0

    valid_percentage_median = 0.0


print(
    f"\nMean daily valid-cell percentage: "
    f"{valid_percentage_mean:.2f}%"
)

print(
    f"Median daily valid-cell percentage: "
    f"{valid_percentage_median:.2f}%"
)


# ======================================================================
# 11. STORAGE SUMMARY
# ======================================================================

original_total_gb = (
    total_original_bytes
    / (1024 ** 3)
)

regional_total_mb = (
    total_regional_bytes
    / (1024 ** 2)
)

regional_total_gb = (
    total_regional_bytes
    / (1024 ** 3)
)


print("\n" + "=" * 70)
print("STORAGE SUMMARY")
print("=" * 70)

print(
    f"Temporary original data processed: "
    f"{original_total_gb:.3f} GB"
)

print(
    f"Final regional CHL storage: "
    f"{regional_total_mb:.3f} MB"
)

print(
    f"Final regional CHL storage: "
    f"{regional_total_gb:.6f} GB"
)

print(
    "\nIMPORTANT:"
)

print(
    "The temporary global files have been deleted."
)


# ======================================================================
# 12. VERIFY NO TEMPORARY GLOBAL FILES REMAIN
# ======================================================================

remaining_nc_files = list(
    TEMP_DIR.rglob(
        "*.nc"
    )
)


print("\n" + "=" * 70)
print("TEMPORARY FILE CHECK")
print("=" * 70)

print(
    "Remaining temporary NetCDF files:",
    len(remaining_nc_files)
)


if len(remaining_nc_files) == 0:

    temporary_cleanup_success = True

    print(
        "✓ No temporary global NetCDF files remain."
    )

else:

    temporary_cleanup_success = False

    print(
        "⚠ Some temporary NetCDF files remain."
    )

    for file_path in remaining_nc_files[:10]:

        print(
            file_path
        )


# ======================================================================
# 13. VERIFY REGIONAL FILE COUNT
# ======================================================================

regional_files = sorted(
    REGIONAL_DIR.glob(
        "*.nc"
    )
)


regional_file_count = len(
    regional_files
)


print("\n" + "=" * 70)
print("FINAL REGIONAL FILE CHECK")
print("=" * 70)

print(
    "Regional CHL files:",
    regional_file_count
)

print(
    "Expected regional files:",
    EXPECTED_DAYS
)


regional_file_count_complete = (
    regional_file_count
    == EXPECTED_DAYS
)


if regional_file_count_complete:

    print(
        "✓ Complete Jan-Apr regional file set created."
    )

else:

    print(
        "⚠ Regional file count is incomplete."
    )


# ======================================================================
# 14. DATE COMPLETENESS CHECK
# ======================================================================

expected_date_strings = {
    d.strftime("%Y%m%d")
    for d in DATES
}


actual_date_strings = set()


for file_path in regional_files:

    filename = file_path.name

    # Extract the YYYYMMDD part
    # from the known filename format.

    parts = filename.split("_")

    for part in parts:

        if (
            len(part) == 8
            and part.isdigit()
            and part.startswith("2026")
        ):

            actual_date_strings.add(
                part
            )

            break


missing_dates = sorted(
    expected_date_strings
    - actual_date_strings
)

unexpected_dates = sorted(
    actual_date_strings
    - expected_date_strings
)


print("\n" + "=" * 70)
print("DATE COMPLETENESS CHECK")
print("=" * 70)

print(
    "Expected dates:",
    len(expected_date_strings)
)

print(
    "Actual regional dates:",
    len(actual_date_strings)
)

print(
    "Missing dates:",
    len(missing_dates)
)

print(
    "Unexpected dates:",
    len(unexpected_dates)
)


if missing_dates:

    print(
        "\nFirst missing dates:"
    )

    for date_value in missing_dates[:20]:

        print(
            date_value
        )


if unexpected_dates:

    print(
        "\nUnexpected dates:"
    )

    for date_value in unexpected_dates[:20]:

        print(
            date_value
        )


date_completeness = (
    len(missing_dates) == 0
    and
    len(unexpected_dates) == 0
)


# ======================================================================
# 15. SAVE FINAL SUMMARY
# ======================================================================

summary_df.to_csv(
    SUMMARY_CSV,
    index=False
)


print(
    "\n✓ Final daily summary saved:"
)

print(
    SUMMARY_CSV
)


# ======================================================================
# 16. FINAL REPORT
# ======================================================================

report = {

    "step":
        "31.9P-16T-24-F",

    "dataset_id":
        DATASET_ID,

    "dataset_version":
        DATASET_VERSION,

    "validation_period": {

        "start":
            START_DATE.strftime(
                "%Y-%m-%d"
            ),

        "end":
            END_DATE.strftime(
                "%Y-%m-%d"
            ),

        "expected_days":
            EXPECTED_DAYS
    },

    "study_region": {

        "latitude_min":
            LAT_MIN,

        "latitude_max":
            LAT_MAX,

        "longitude_min":
            LON_MIN,

        "longitude_max":
            LON_MAX
    },

    "processing": {

        "successful_days":
            successful_days,

        "failed_days":
            failed_days,

        "days_with_valid_chl":
            days_with_valid_chl,

        "days_with_no_valid_chl":
            days_with_no_valid_chl
    },

    "coverage": {

        "mean_daily_valid_percentage":
            float(
                valid_percentage_mean
            ),

        "median_daily_valid_percentage":
            float(
                valid_percentage_median
            )
    },

    "storage": {

        "temporary_original_data_gb":
            float(
                original_total_gb
            ),

        "final_regional_data_mb":
            float(
                regional_total_mb
            ),

        "final_regional_data_gb":
            float(
                regional_total_gb
            )
    },

    "files": {

        "regional_file_count":
            regional_file_count,

        "expected_file_count":
            EXPECTED_DAYS,

        "date_completeness":
            date_completeness,

        "missing_dates":
            missing_dates,

        "unexpected_dates":
            unexpected_dates,

        "temporary_global_files_remaining":
            len(
                remaining_nc_files
            )
    },

    "scientific_safety": {

        "xgboost_modified":
            False,

        "xgboost_retrained":
            False,

        "predictions_regenerated":
            False,

        "shap_modified":
            False,

        "three_d_visualization_modified":
            False,

        "missing_chl_filled":
            False
    },

    "output": {

        "regional_directory":
            str(
                REGIONAL_DIR
            ),

        "summary_csv":
            str(
                SUMMARY_CSV
            ),

        "report_json":
            str(
                REPORT_JSON
            )
    }
}


with open(
    REPORT_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        report,
        f,
        indent=2
    )


# ======================================================================
# 17. FINAL SAFETY CHECK
# ======================================================================

print("\n" + "=" * 70)
print("FINAL 31.9P-16T-24-F SAFETY CHECK")
print("=" * 70)

print(
    f"120 calendar days processed        : "
    f"{len(summary_df) == EXPECTED_DAYS}"
)

print(
    f"Regional files complete            : "
    f"{regional_file_count_complete}"
)

print(
    f"Dates complete                     : "
    f"{date_completeness}"
)

print(
    f"Temporary global files removed     : "
    f"{temporary_cleanup_success}"
)

print(
    f"Valid CHL days identified          : "
    f"{days_with_valid_chl}"
)

print(
    "Missing CHL artificially filled    : FALSE"
)

print(
    "XGBoost modified                   : FALSE"
)

print(
    "XGBoost retrained                  : FALSE"
)

print(
    "Predictions regenerated            : FALSE"
)

print(
    "SHAP modified                      : FALSE"
)

print(
    "3D visualization modified          : FALSE"
)


# ======================================================================
# 18. COMPLETE
# ======================================================================

print("\n" + "=" * 70)
print("STEP 31.9P-16T-24-F COMPLETE")
print("=" * 70)

print(
    "\nRegional CHL directory:"
)

print(
    REGIONAL_DIR
)

print(
    "\nDaily summary:"
)

print(
    SUMMARY_CSV
)

print(
    "\nVerification report:"
)

print(
    REPORT_JSON
)

print(
    "\nIMPORTANT:"
)

print(
    "Do NOT perform external model validation yet."
)

print(
    "Send me the COMPLETE output first."
)
```
