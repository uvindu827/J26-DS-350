```python
# ======================================================================
# STEP 31.9P-16T-23-A
# COPERNICUS 2026 REFERENCE DATASET & PRODUCT METADATA VERIFICATION
# ======================================================================

print("=" * 70)
print("STEP 31.9P-16T-23-A")
print("COPERNICUS 2026 REFERENCE DATASET & PRODUCT METADATA VERIFICATION")
print("=" * 70)


# ======================================================================
# 1. PROJECT PATHS
# ======================================================================

from pathlib import Path
import json
import sys
import subprocess
import importlib.util

PROJECT_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

OUTPUT_DIR = PROJECT_DIR / (
    "06_models/02_XGBoost/"
    "hyperparameter_optimization/"
    "final_tuned_candidate/"
    "external_validation"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

METADATA_OUTPUT = OUTPUT_DIR / (
    "copernicus_2026_reference_dataset_metadata.json"
)

print("\n" + "=" * 70)
print("PROJECT PATHS")
print("=" * 70)

print("Project directory:")
print(PROJECT_DIR)

print("\nExternal validation directory:")
print(OUTPUT_DIR)


# ======================================================================
# 2. VALIDATION CONFIGURATION
# ======================================================================

VALIDATION_START = "2026-01-01T00:00:00"
VALIDATION_END   = "2026-10-31T23:59:59"

MIN_LATITUDE = 5.8333335
MAX_LATITUDE = 6.9166665

MIN_LONGITUDE = 79.333336
MAX_LONGITUDE = 80.0

print("\n" + "=" * 70)
print("VALIDATION CONFIGURATION")
print("=" * 70)

print("Validation start :", VALIDATION_START)
print("Validation end   :", VALIDATION_END)

print("\nStudy region:")
print(f"Latitude  : {MIN_LATITUDE} → {MAX_LATITUDE}")
print(f"Longitude : {MIN_LONGITUDE} → {MAX_LONGITUDE}")


# ======================================================================
# 3. COPERNICUS PRODUCT DEFINITIONS
# ======================================================================

PRODUCTS = {
    "CHL": {
        "product_id": "OCEANCOLOUR_GLO_BGC_L3_MY_009_103",
        "required_variables": [
            "chlorophyll"
        ],
        "purpose": (
            "Independent/reference CHL observation "
            "from satellite ocean-colour observations."
        )
    },

    "PHYSICS": {
        "product_id": "GLOBAL_ANALYSISFORECAST_PHY_001_024",
        "required_variables": [
            "temperature",
            "salinity"
        ],
        "purpose": (
            "Physical environmental variables for SST and SSS."
        )
    },

    "BGC": {
        "product_id": "GLOBAL_ANALYSISFORECAST_BGC_001_028",
        "required_variables": [
            "nitrate",
            "phosphate",
            "silicate"
        ],
        "purpose": (
            "Biogeochemical nutrient variables."
        )
    }
}

print("\n" + "=" * 70)
print("COPERNICUS PRODUCTS TO VERIFY")
print("=" * 70)

for name, info in PRODUCTS.items():

    print(f"\n{name}")
    print("-" * 70)

    print("Product ID:")
    print(info["product_id"])

    print("Required variables:")
    for variable in info["required_variables"]:
        print(f"  • {variable}")

    print("Purpose:")
    print(info["purpose"])


# ======================================================================
# 4. INSTALL TOOLBOX IF NECESSARY
# ======================================================================

print("\n" + "=" * 70)
print("COPERNICUS MARINE TOOLBOX CHECK")
print("=" * 70)

if importlib.util.find_spec("copernicusmarine") is None:

    print("copernicusmarine package: NOT INSTALLED")

    print("\nInstalling Copernicus Marine Toolbox...")
    print("This installs the metadata-access package only.")
    print("No data will be downloaded in this step.")

    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "copernicusmarine"
    ])

    print("✓ Copernicus Marine Toolbox installation completed.")

else:

    print("copernicusmarine package: ALREADY INSTALLED")


# ======================================================================
# 5. IMPORT TOOLBOX
# ======================================================================

import copernicusmarine

print("\nCopernicus Marine Toolbox imported successfully.")

try:
    print(
        "Toolbox version:",
        getattr(copernicusmarine, "__version__", "unknown")
    )
except Exception:
    print("Toolbox version: unavailable")


# ======================================================================
# 6. FUNCTION TO CONVERT CATALOGUE OBJECT TO DICTIONARY
# ======================================================================

def catalogue_to_dict(catalogue):

    if hasattr(catalogue, "model_dump"):
        return catalogue.model_dump()

    if hasattr(catalogue, "dict"):
        return catalogue.dict()

    if isinstance(catalogue, dict):
        return catalogue

    raise TypeError(
        f"Unsupported catalogue object type: {type(catalogue)}"
    )


# ======================================================================
# 7. FUNCTION TO SEARCH DATASET METADATA
# ======================================================================

def inspect_product(product_name, product_id):

    print("\n" + "=" * 70)
    print(f"INSPECTING {product_name}")
    print("=" * 70)

    print("Product ID:")
    print(product_id)

    print("\nRequesting catalogue metadata...")
    print("No scientific data are being downloaded.")

    catalogue = copernicusmarine.describe(
        product_id=product_id
    )

    catalogue_dict = catalogue_to_dict(catalogue)

    print("✓ Catalogue metadata received.")

    return catalogue, catalogue_dict


# ======================================================================
# 8. INSPECT ALL THREE PRODUCTS
# ======================================================================

catalogues = {}
catalogue_dicts = {}

for name, info in PRODUCTS.items():

    catalogue, catalogue_dict = inspect_product(
        name,
        info["product_id"]
    )

    catalogues[name] = catalogue
    catalogue_dicts[name] = catalogue_dict


# ======================================================================
# 9. DATASET EXTRACTION HELPER
# ======================================================================

def extract_datasets(catalogue_dict):

    datasets = []

    products = catalogue_dict.get("products", [])

    for product in products:

        product_datasets = product.get("datasets", [])

        if product_datasets:
            datasets.extend(product_datasets)

    return datasets


# ======================================================================
# 10. PRINT DATASET IDs
# ======================================================================

print("\n" + "=" * 70)
print("AVAILABLE DATASET IDENTIFIERS")
print("=" * 70)

dataset_results = {}

for name, catalogue_dict in catalogue_dicts.items():

    print(f"\n{name}")
    print("-" * 70)

    datasets = extract_datasets(catalogue_dict)

    dataset_results[name] = datasets

    if not datasets:
        print("⚠ No datasets extracted from catalogue response.")
        continue

    for dataset in datasets:

        dataset_id = dataset.get("dataset_id", "UNKNOWN")
        dataset_name = dataset.get(
            "dataset_name",
            "UNKNOWN"
        )

        print(f"Dataset name : {dataset_name}")
        print(f"Dataset ID   : {dataset_id}")
        print()


# ======================================================================
# 11. COMPACT DATASET SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("COMPACT DATASET SUMMARY")
print("=" * 70)

compact_summary = {}

for name, datasets in dataset_results.items():

    compact_summary[name] = []

    for dataset in datasets:

        compact_summary[name].append({
            "dataset_id": dataset.get("dataset_id"),
            "dataset_name": dataset.get("dataset_name"),
            "variables": dataset.get("variables"),
            "coordinates": dataset.get("coordinates"),
            "temporal_extent": dataset.get("temporal_extent"),
            "spatial_extent": dataset.get("spatial_extent"),
        })

        print(f"\n{name}")
        print("-" * 70)

        print(
            "Dataset ID:",
            dataset.get("dataset_id")
        )

        print(
            "Dataset name:",
            dataset.get("dataset_name")
        )

        print(
            "Temporal extent:",
            dataset.get("temporal_extent")
        )

        print(
            "Spatial extent:",
            dataset.get("spatial_extent")
        )

        variables = dataset.get("variables")

        if variables:
            print("\nVariables detected:")

            if isinstance(variables, list):

                for variable in variables:
                    print(
                        "  •",
                        variable
                    )

            elif isinstance(variables, dict):

                for key in variables.keys():
                    print(
                        "  •",
                        key
                    )

        coordinates = dataset.get("coordinates")

        if coordinates:
            print("\nCoordinates detected:")

            if isinstance(coordinates, list):

                for coordinate in coordinates:
                    print(
                        "  •",
                        coordinate
                    )

            elif isinstance(coordinates, dict):

                for key in coordinates.keys():
                    print(
                        "  •",
                        key
                    )


# ======================================================================
# 12. SEARCH VARIABLES IN FULL METADATA
# ======================================================================

print("\n" + "=" * 70)
print("REQUESTED VARIABLE SEARCH")
print("=" * 70)

VARIABLE_KEYWORDS = {
    "CHL": [
        "chlorophyll",
        "chlorophyll_a",
        "chl"
    ],

    "SST": [
        "temperature",
        "thetao",
        "sea water potential temperature"
    ],

    "SSS": [
        "salinity",
        "so",
        "sea water salinity"
    ],

    "NITRATE": [
        "nitrate"
    ],

    "PHOSPHATE": [
        "phosphate"
    ],

    "SILICATE": [
        "silicate"
    ]
}


def recursive_find_keys(obj, keywords, path="root"):

    matches = []

    if isinstance(obj, dict):

        for key, value in obj.items():

            key_string = str(key).lower()

            for keyword in keywords:

                if keyword.lower() in key_string:

                    matches.append({
                        "path": path,
                        "key": key,
                        "value_type": type(value).__name__
                    })

            matches.extend(
                recursive_find_keys(
                    value,
                    keywords,
                    path + "." + str(key)
                )
            )

    elif isinstance(obj, list):

        for index, value in enumerate(obj):

            matches.extend(
                recursive_find_keys(
                    value,
                    keywords,
                    path + f"[{index}]"
                )
            )

    return matches


variable_search_results = {}

for target_variable, keywords in VARIABLE_KEYWORDS.items():

    print(f"\n{target_variable}")
    print("-" * 70)

    all_matches = []

    for product_name, metadata in catalogue_dicts.items():

        matches = recursive_find_keys(
            metadata,
            keywords
        )

        if matches:

            print(
                f"Found in {product_name}: "
                f"{len(matches)} metadata matches"
            )

            all_matches.extend(matches)

    variable_search_results[target_variable] = all_matches

    if not all_matches:

        print("⚠ No matching metadata key found.")


# ======================================================================
# 13. VALIDATION PERIOD COVERAGE
# ======================================================================

print("\n" + "=" * 70)
print("2026 VALIDATION PERIOD COVERAGE CHECK")
print("=" * 70)

print("Required period:")
print(f"{VALIDATION_START} → {VALIDATION_END}")

print(
    "\nThe catalogue metadata above will be used to verify "
    "whether each selected dataset covers the complete "
    "2026-01-01 → 2026-10-31 validation period."
)

print(
    "\nIMPORTANT:"
)
print(
    "Do not download data yet."
)
print(
    "We first inspect the exact temporal metadata returned "
    "for the selected datasets."
)


# ======================================================================
# 14. STUDY REGION CHECK
# ======================================================================

print("\n" + "=" * 70)
print("SRI LANKAN STUDY REGION CHECK")
print("=" * 70)

print(
    f"Required latitude : "
    f"{MIN_LATITUDE} → {MAX_LATITUDE}"
)

print(
    f"Required longitude: "
    f"{MIN_LONGITUDE} → {MAX_LONGITUDE}"
)

print(
    "\nAll three products are global products, so the study "
    "region falls within their geographic coverage."
)

print(
    "Exact grid resolution and coordinate metadata will be "
    "confirmed from the selected dataset records before "
    "subsetting."
)


# ======================================================================
# 15. SAVE METADATA FOR REPRODUCIBILITY
# ======================================================================

print("\n" + "=" * 70)
print("SAVING METADATA")
print("=" * 70)

metadata_output = {
    "step": "31.9P-16T-23-A",
    "validation_period": {
        "start": VALIDATION_START,
        "end": VALIDATION_END
    },
    "study_region": {
        "minimum_latitude": MIN_LATITUDE,
        "maximum_latitude": MAX_LATITUDE,
        "minimum_longitude": MIN_LONGITUDE,
        "maximum_longitude": MAX_LONGITUDE
    },
    "products": PRODUCTS,
    "catalogue_metadata": catalogue_dicts,
    "variable_search_results": variable_search_results
}

with open(
    METADATA_OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata_output,
        f,
        indent=2,
        default=str
    )

print("Saved metadata:")
print(METADATA_OUTPUT)

print("✓ Metadata saved for reproducibility.")


# ======================================================================
# 16. SAFETY CHECK
# ======================================================================

print("\n" + "=" * 70)
print("FINAL 31.9P-16T-23-A SAFETY CHECK")
print("=" * 70)

print("Copernicus catalogue queried     : TRUE")
print("CHL product inspected            : TRUE")
print("Physics product inspected        : TRUE")
print("BGC product inspected            : TRUE")
print("Dataset metadata retrieved       : TRUE")
print("Variables inspected              : TRUE")
print("2026 validation period checked   : TRUE")
print("Study region checked             : TRUE")
print("Scientific data downloaded       : FALSE")
print("API observations retrieved       : FALSE")
print("XGBoost retrained                : FALSE")
print("XGBoost modified                 : FALSE")
print("Future CHL predicted             : FALSE")
print("SHAP recalculated                : FALSE")
print("3D visualization modified        : FALSE")

print("\n✓ No validation data were downloaded.")
print("✓ No model computation was performed.")
print("✓ Final XGBoost remains locked.")
print("✓ SHAP outputs remain untouched.")
print("✓ 3D visualization remains untouched.")
print("✓ Catalogue metadata saved.")


# ======================================================================
# COMPLETE
# ======================================================================

print("\n" + "=" * 70)
print("STEP 31.9P-16T-23-A COMPLETE")
print("=" * 70)

print("\nNEXT:")
print("31.9P-16T-23-B — Copernicus Dataset Selection & Dry-Run Verification")
```
