```python
import xarray as xr
import numpy as np
import pandas as pd

# ============================================================
# STEP 1 — OPEN ALL THREE DATASETS
# ============================================================

phy = xr.open_dataset(
    "copernicus_data/SST_SSS_Western_Southern_SriLanka_daily.nc"
)

bgc = xr.open_dataset(
    "copernicus_data/Nutrients_Western_Southern_SriLanka_daily.nc"
)

chl = xr.open_dataset(
    "copernicus_data/Chla_Western_Southern_SriLanka_daily.nc"
)

print("=" * 70)
print("STEP 1 — DATASET INFORMATION")
print("=" * 70)

print("\n========== PHYSICAL DATA (SST + SSS) ==========")
print(phy)

print("\n========== NUTRIENT DATA ==========")
print(bgc)

print("\n========== CHLOROPHYLL-a DATA ==========")
print(chl)


# ============================================================
# STEP 2 — CHECK SPATIAL RESOLUTION
# ============================================================

print("\n" + "=" * 70)
print("STEP 2 — SPATIAL GRID INFORMATION")
print("=" * 70)

print("\nPHYSICAL DATA")
print("Latitude points :", len(phy.latitude))
print("Longitude points:", len(phy.longitude))

print("\nNUTRIENT DATA")
print("Latitude points :", len(bgc.latitude))
print("Longitude points:", len(bgc.longitude))

print("\nCHLOROPHYLL DATA")
print("Latitude points :", len(chl.latitude))
print("Longitude points:", len(chl.longitude))

print("\nPhysical latitude range:")
print(float(phy.latitude.min()), "to", float(phy.latitude.max()))

print("Physical longitude range:")
print(float(phy.longitude.min()), "to", float(phy.longitude.max()))

print("\nNutrient latitude range:")
print(float(bgc.latitude.min()), "to", float(bgc.latitude.max()))

print("Nutrient longitude range:")
print(float(bgc.longitude.min()), "to", float(bgc.longitude.max()))

print("\nChlorophyll latitude range:")
print(float(chl.latitude.min()), "to", float(chl.latitude.max()))

print("Chlorophyll longitude range:")
print(float(chl.longitude.min()), "to", float(chl.longitude.max()))


# ============================================================
# STEP 3 — CHECK TIME RANGES
# ============================================================

print("\n" + "=" * 70)
print("STEP 3 — TEMPORAL COVERAGE")
print("=" * 70)

print("\nPhysical:")
print(phy.time.min().values, "to", phy.time.max().values)

print("\nNutrients:")
print(bgc.time.min().values, "to", bgc.time.max().values)

print("\nChlorophyll:")
print(chl.time.min().values, "to", chl.time.max().values)


# ============================================================
# STEP 4 — CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 4 — MISSING VALUE ANALYSIS")
print("=" * 70)

for name, ds in [
    ("Physical", phy),
    ("Nutrients", bgc),
    ("Chlorophyll", chl)
]:

    print(f"\n{name}")

    for var in ds.data_vars:

        total = ds[var].size
        missing = int(ds[var].isnull().sum())
        percentage = (missing / total) * 100

        print(
            f"{var}: "
            f"{missing:,} missing / "
            f"{total:,} total "
            f"({percentage:.2f}%)"
        )


# ============================================================
# STEP 5 — CHECK VARIABLE UNITS
# ============================================================

print("\n" + "=" * 70)
print("STEP 5 — VARIABLE INFORMATION AND UNITS")
print("=" * 70)

print("\n========== PHYSICAL VARIABLES ==========")

for var in phy.data_vars:
    print(
        var,
        "| units:", phy[var].attrs.get("units"),
        "| long_name:", phy[var].attrs.get("long_name")
    )

print("\n========== NUTRIENT VARIABLES ==========")

for var in bgc.data_vars:
    print(
        var,
        "| units:", bgc[var].attrs.get("units"),
        "| long_name:", bgc[var].attrs.get("long_name")
    )

print("\n========== CHLOROPHYLL ==========")

for var in chl.data_vars:
    print(
        var,
        "| units:", chl[var].attrs.get("units"),
        "| long_name:", chl[var].attrs.get("long_name")
    )


# ============================================================
# STEP 6 — CHECK VALUE RANGES
# ============================================================

print("\n" + "=" * 70)
print("STEP 6 — VARIABLE VALUE RANGES")
print("=" * 70)

def show_ranges(ds, name):

    print(f"\n========== {name} ==========")

    for var in ds.data_vars:

        values = ds[var]

        print(
            f"{var}: "
            f"min={float(values.min(skipna=True)):.6f}, "
            f"max={float(values.max(skipna=True)):.6f}, "
            f"mean={float(values.mean(skipna=True)):.6f}"
        )


show_ranges(phy, "PHYSICAL")
show_ranges(bgc, "NUTRIENTS")
show_ranges(chl, "CHLOROPHYLL")


# ============================================================
# STEP 7 — CHECK DEPTH INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 7 — DEPTH INFORMATION")
print("=" * 70)

print("\nPhysical depth:")
print(phy.depth.values)

print("\nNutrient depth:")
print(bgc.depth.values)

print("\nChlorophyll coordinates:")
print(chl.coords)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("INITIAL DATA INSPECTION COMPLETED")
print("=" * 70)

print("\nDatasets successfully loaded:")
print("✓ SST + SSS")
print("✓ Nitrate + Phosphate + Silicate")
print("✓ Historical Chlorophyll-a")

print("\nResearch region:")
print("Western + Southern Sri Lankan coastal waters")

print("\nNext stage:")
print("Spatial-temporal alignment and integrated dataset preparation")
```

```python
target_lat = bgc.latitude
target_lon = bgc.longitude

print("Target latitude points :", len(target_lat))
print("Target longitude points:", len(target_lon))
```

```python
# ============================================================
# STEP 11 — REGRID SST + SSS TO THE COMMON NUTRIENT GRID
# ============================================================

phy_regridded = phy.interp(
    latitude=target_lat,
    longitude=target_lon,
    method="linear"
)

print("Regridded physical dataset:")
print(phy_regridded)

print("\nVariables:")
print(list(phy_regridded.data_vars))

print("\nNew spatial dimensions:")
print("Latitude points :", len(phy_regridded.latitude))
print("Longitude points:", len(phy_regridded.longitude))
```

```python
# ============================================================
# STEP 12 — REGRID CHLOROPHYLL-a TO THE COMMON GRID
# ============================================================

chl_regridded = chl.interp(
    latitude=target_lat,
    longitude=target_lon,
    method="linear"
)

print("Regridded Chlorophyll-a dataset:")
print(chl_regridded)

print("\nVariables:")
print(list(chl_regridded.data_vars))

print("\nNew spatial dimensions:")
print("Latitude points :", len(chl_regridded.latitude))
print("Longitude points:", len(chl_regridded.longitude))
```

```python
# ============================================================
# STEP 13 — CHECK TEMPORAL ALIGNMENT
# ============================================================

print("SST/SSS time points :", len(phy_regridded.time))
print("Nutrient time points:", len(bgc.time))
print("Chl-a time points   :", len(chl_regridded.time))

print("\nFirst dates:")
print("Physical :", phy_regridded.time.values[0])
print("Nutrients:", bgc.time.values[0])
print("Chl-a    :", chl_regridded.time.values[0])

print("\nLast dates:")
print("Physical :", phy_regridded.time.values[-1])
print("Nutrients:", bgc.time.values[-1])
print("Chl-a    :", chl_regridded.time.values[-1])


# Check whether all time coordinates are exactly identical
physical_times = phy_regridded.time.values
nutrient_times = bgc.time.values
chl_times = chl_regridded.time.values

print("\nAre Physical and Nutrient dates identical?")
print(np.array_equal(physical_times, nutrient_times))

print("\nAre Physical and Chl-a dates identical?")
print(np.array_equal(physical_times, chl_times))

print("\nAre Nutrient and Chl-a dates identical?")
print(np.array_equal(nutrient_times, chl_times))
```

```python
# ============================================================
# STEP 14 — MERGE THE THREE ALIGNED DATASETS
# ============================================================

combined = xr.merge([
    phy_regridded,
    bgc,
    chl_regridded
])

print("========== COMBINED DATASET ==========")
print(combined)

print("\nVariables:")
print(list(combined.data_vars))

print("\nDimensions:")
print(combined.dims)
```

```python
# ============================================================
# STEP 14A — REMOVE DEPTH DIMENSION
# Surface-level dataset
# ============================================================

# Select the only available surface depth
phy_surface = phy_regridded.isel(depth=0).drop_vars("depth")
bgc_surface = bgc.isel(depth=0).drop_vars("depth")

print("Physical dataset after removing depth:")
print(phy_surface)

print("\nNutrient dataset after removing depth:")
print(bgc_surface)
```

```python
# ============================================================
# STEP 14B — MERGE AFTER DEPTH CORRECTION
# ============================================================

combined = xr.merge(
    [
        phy_surface,
        bgc_surface,
        chl_regridded
    ],
    join="exact"
)

print("========== CORRECTED COMBINED DATASET ==========")
print(combined)

print("\nVariables:")
print(list(combined.data_vars))

print("\nDimensions:")
print(combined.dims)
```

```python
# ============================================================
# STEP 15 — MISSING VALUES AFTER SPATIAL ALIGNMENT
# ============================================================

print("=" * 70)
print("MISSING VALUES IN FINAL COMMON GRID")
print("=" * 70)

for var in combined.data_vars:

    total = combined[var].size
    missing = int(combined[var].isnull().sum())
    valid = total - missing
    percentage = (missing / total) * 100
    valid_percentage = (valid / total) * 100

    print(f"\n{var}")
    print(f"Total values   : {total:,}")
    print(f"Valid values   : {valid:,}")
    print(f"Missing values : {missing:,}")
    print(f"Valid %        : {valid_percentage:.2f}%")
    print(f"Missing %      : {percentage:.2f}%")
```

```python
# ============================================================
# STEP 15B — MISSING VALUES BY SPATIAL LOCATION
# ============================================================

for var in combined.data_vars:

    print("\n" + "=" * 70)
    print(f"{var.upper()} — SPATIAL VALIDITY")
    print("=" * 70)

    spatial_missing = combined[var].isnull().mean(dim="time") * 100

    print(spatial_missing)
```

```python
# ============================================================
# STEP 15C — IDENTIFY COMPLETELY MISSING GRID CELLS
# ============================================================

for var in combined.data_vars:

    all_missing = combined[var].isnull().all(dim="time")

    print("\n" + "=" * 70)
    print(f"{var.upper()} — COMPLETELY MISSING CELLS")
    print("=" * 70)

    print(all_missing)
```

```python
# ============================================================
# STEP 16 — FIND COMMON VALID SPATIAL CELLS
# ============================================================

variables = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate",
    "chlorophyll_a"
]

# A cell is considered spatially valid if every variable
# has at least one valid observation during the study period.

valid_masks = []

for var in variables:
    valid_masks.append(
        combined[var].notnull().any(dim="time")
    )

common_valid_cells = valid_masks[0]

for mask in valid_masks[1:]:
    common_valid_cells = common_valid_cells & mask

print("=" * 70)
print("COMMON VALID SPATIAL CELLS")
print("=" * 70)

print(common_valid_cells)

print("\nNumber of common valid cells:",
      int(common_valid_cells.sum()))

print("Total spatial cells:",
      common_valid_cells.size)

print("Percentage of spatial grid:",
      float(common_valid_cells.mean() * 100))
```

```python
# ============================================================
# STEP 16B — LIST COMMON VALID GRID LOCATIONS
# ============================================================

valid_locations = []

for lat in combined.latitude.values:
    for lon in combined.longitude.values:

        is_valid = bool(
            common_valid_cells.sel(
                latitude=lat,
                longitude=lon
            ).values
        )

        if is_valid:
            valid_locations.append(
                (float(lat), float(lon))
            )

print("=" * 70)
print("VALID COMMON LOCATIONS")
print("=" * 70)

for lat, lon in valid_locations:
    print(
        f"Latitude: {lat:.2f} | "
        f"Longitude: {lon:.2f}"
    )
```

```python
# ============================================================
# STEP 17 — ORIGINAL CHL-a SPATIAL COVERAGE
# ============================================================

print("=" * 70)
print("ORIGINAL CHLOROPHYLL-a SPATIAL COVERAGE")
print("=" * 70)

chl_original_valid = chl["chlorophyll_a"].notnull().mean(dim="time") * 100

print("Original Chl-a grid:")
print("Latitude points :", len(chl.latitude))
print("Longitude points:", len(chl.longitude))

print("\nLatitude range:")
print(float(chl.latitude.min()), "to", float(chl.latitude.max()))

print("\nLongitude range:")
print(float(chl.longitude.min()), "to", float(chl.longitude.max()))

print("\nValid observation percentage by original grid cell:")
print(chl_original_valid)
```

```python
# ============================================================
# STEP 17B — COUNT ORIGINAL CHL-a VALID CELLS
# ============================================================

original_chl_valid_cells = (
    chl["chlorophyll_a"]
    .notnull()
    .any(dim="time")
)

print("=" * 70)
print("ORIGINAL CHL-a VALID CELL COUNT")
print("=" * 70)

print(
    "Valid original Chl-a cells:",
    int(original_chl_valid_cells.sum())
)

print(
    "Total original Chl-a cells:",
    original_chl_valid_cells.size
)

print(
    "Percentage of original grid:",
    float(original_chl_valid_cells.mean() * 100)
)
```

```python
# ============================================================
# STEP 19C — COMMON VALID CELLS AT 0.083°
# ============================================================

valid_masks_0083 = [
    phy_0083["sst"].notnull().any(dim="time"),
    phy_0083["sss"].notnull().any(dim="time"),
    bgc_0083["nitrate"].notnull().any(dim="time"),
    bgc_0083["phosphate"].notnull().any(dim="time"),
    bgc_0083["silicate"].notnull().any(dim="time"),
    chl_0083["chlorophyll_a"].notnull().any(dim="time")
]

common_valid_0083 = valid_masks_0083[0]

for mask in valid_masks_0083[1:]:
    common_valid_0083 = common_valid_0083 & mask

print("=" * 70)
print("0.083° COMMON VALID GRID")
print("=" * 70)

print("Total spatial cells:",
      common_valid_0083.size)

print("Common valid cells:",
      int(common_valid_0083.sum()))

print("Percentage:",
      float(common_valid_0083.mean() * 100))
```

```python
# ============================================================
# STEP 20A — COMBINE THE 0.083° DATASETS
# ============================================================

combined_0083 = xr.merge(
    [
        phy_0083,
        bgc_0083,
        chl_0083
    ],
    join="exact"
)

print("=" * 70)
print("0.083° COMBINED DATASET")
print("=" * 70)

print(combined_0083)

print("\nVariables:")
print(list(combined_0083.data_vars))

print("\nDimensions:")
print(combined_0083.sizes)
```

```python
# ============================================================
# STEP 20B — TEMPORAL COVERAGE PER VARIABLE
# ============================================================

variables = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate",
    "chlorophyll_a"
]

coverage = {}

for var in variables:

    coverage[var] = (
        combined_0083[var]
        .notnull()
        .mean(dim="time") * 100
    )

    print("\n" + "=" * 70)
    print(f"{var.upper()} TEMPORAL COVERAGE (%)")
    print("=" * 70)

    print(coverage[var])
```

```python
# ============================================================
# STEP 20C — SUMMARY OF TEMPORAL COVERAGE
# ============================================================

print("=" * 70)
print("TEMPORAL COVERAGE SUMMARY")
print("=" * 70)

for var in variables:

    values = coverage[var].values

    valid_cells = np.isfinite(values)

    print(f"\n{var}")
    print(f"Minimum coverage : {np.nanmin(values):.2f}%")
    print(f"Maximum coverage : {np.nanmax(values):.2f}%")
    print(f"Mean coverage    : {np.nanmean(values):.2f}%")
```

```python
# ============================================================
# STEP 20D — MINIMUM MULTI-VARIABLE TEMPORAL COVERAGE
# ============================================================

coverage_stack = xr.concat(
    [
        coverage["sst"],
        coverage["sss"],
        coverage["nitrate"],
        coverage["phosphate"],
        coverage["silicate"],
        coverage["chlorophyll_a"]
    ],
    dim="variable"
)

minimum_coverage = coverage_stack.min(dim="variable")

print("=" * 70)
print("MINIMUM TEMPORAL COVERAGE ACROSS ALL SIX VARIABLES")
print("=" * 70)

print(minimum_coverage)
```

```python
# ============================================================
# STEP 20E — COMPARE TEMPORAL COVERAGE THRESHOLDS
# ============================================================

thresholds = [50, 60, 70, 80, 90]

print("=" * 70)
print("TEMPORAL COVERAGE THRESHOLD ANALYSIS")
print("=" * 70)

total_cells = minimum_coverage.size

for threshold in thresholds:

    valid_cells = minimum_coverage >= threshold

    count = int(valid_cells.sum())

    percentage = (count / total_cells) * 100

    print(
        f"\nThreshold: {threshold}%"
        f"\nValid cells: {count}"
        f"\nSpatial coverage: {percentage:.2f}%"
    )
```

```python
# ============================================================
# STEP 21 — FINAL COMMON VALID SPATIAL MASK
# ============================================================

variables = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate",
    "chlorophyll_a"
]

# A cell is valid only if every variable has
# at least one valid observation through the study period.

valid_masks = []

for var in variables:
    valid_masks.append(
        combined_0083[var].notnull().any(dim="time")
    )

common_valid_mask = valid_masks[0]

for mask in valid_masks[1:]:
    common_valid_mask = common_valid_mask & mask

print("=" * 70)
print("FINAL COMMON VALID SPATIAL MASK")
print("=" * 70)

print(common_valid_mask)

print("\nTotal spatial cells:",
      common_valid_mask.size)

print("Common valid cells:",
      int(common_valid_mask.sum()))

print(
    "Spatial coverage:",
    f"{(common_valid_mask.sum().item() / common_valid_mask.size) * 100:.2f}%"
)
```

```python
# ============================================================
# STEP 31.9E — SEASONAL CHLOROPHYLL-a ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

print("=" * 70)
print("STEP 31.9E — SEASONAL CHLOROPHYLL-a ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. Define seasons
# ------------------------------------------------------------

def assign_season(month):

    if month in [12, 1, 2]:
        return "Northeast Monsoon"

    elif month in [3, 4]:
        return "First Inter-monsoon"

    elif month in [5, 6, 7, 8, 9]:
        return "Southwest Monsoon"

    elif month in [10, 11]:
        return "Second Inter-monsoon"


df_final["season"] = df_final["month"].apply(assign_season)


# ------------------------------------------------------------
# 2. Set the correct seasonal order
# ------------------------------------------------------------

season_order = [
    "Northeast Monsoon",
    "First Inter-monsoon",
    "Southwest Monsoon",
    "Second Inter-monsoon"
]

df_final["season"] = pd.Categorical(
    df_final["season"],
    categories=season_order,
    ordered=True
)


# ------------------------------------------------------------
# 3. Calculate seasonal statistics
# ------------------------------------------------------------

seasonal_chla = (
    df_final
    .groupby("season", observed=True)["chlorophyll_a"]
    .agg(
        count="count",
        mean="mean",
        median="median",
        std="std",
        min="min",
        max="max"
    )
)

print("\n" + "=" * 70)
print("SEASONAL CHLOROPHYLL-a STATISTICS")
print("=" * 70)

print(seasonal_chla.round(6))


# ------------------------------------------------------------
# 4. Plot seasonal mean chlorophyll-a
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    seasonal_chla.index,
    seasonal_chla["mean"]
)

plt.title(
    "Seasonal Mean Chlorophyll-a in Western and Southern "
    "Sri Lankan Coastal Waters",
    fontsize=14
)

plt.xlabel("Season")
plt.ylabel("Mean Chlorophyll-a (mg/m³)")

plt.xticks(
    rotation=20,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


# ------------------------------------------------------------
# 5. Save figure
# ------------------------------------------------------------

eda_folder = (
    "/content/drive/MyDrive/"
    "Indhi_Marine_Research_Data/03_eda"
)

os.makedirs(
    eda_folder,
    exist_ok=True
)

seasonal_plot_path = os.path.join(
    eda_folder,
    "seasonal_chlorophyll_a.png"
)

plt.savefig(
    seasonal_plot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nFigure saved to:")
print(seasonal_plot_path)


# ------------------------------------------------------------
# 6. Save seasonal statistics
# ------------------------------------------------------------

seasonal_stats_path = os.path.join(
    eda_folder,
    "seasonal_chlorophyll_a_statistics.csv"
)

seasonal_chla.to_csv(
    seasonal_stats_path
)

print("\nSeasonal statistics saved to:")
print(seasonal_stats_path)
```

```python
# ============================================================
# STEP 31.9I — TEMPORAL RELATIONSHIPS WITH CHLOROPHYLL-a
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

print("=" * 70)
print("STEP 31.9I — TEMPORAL RELATIONSHIPS WITH CHLOROPHYLL-a")
print("=" * 70)


# ------------------------------------------------------------
# 1. Variables
# ------------------------------------------------------------

environmental_variables = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate"
]

target = "chlorophyll_a"


# ------------------------------------------------------------
# 2. Use yearly environmental means calculated in Step 31.9H
# ------------------------------------------------------------

yearly_data = yearly_environmental.copy()


print("\nNumber of yearly observations:")
print(len(yearly_data))

print("\nYears:")
print(
    yearly_data["year"].min(),
    "to",
    yearly_data["year"].max()
)


# ------------------------------------------------------------
# 3. Pearson correlations with yearly chlorophyll-a
# ------------------------------------------------------------

temporal_correlations = {}

for variable in environmental_variables:

    correlation = yearly_data[
        variable
    ].corr(
        yearly_data[target],
        method="pearson"
    )

    temporal_correlations[variable] = correlation


temporal_corr_df = (
    pd.DataFrame.from_dict(
        temporal_correlations,
        orient="index",
        columns=["pearson_r"]
    )
    .sort_values(
        "pearson_r",
        ascending=False
    )
)


print("\n" + "=" * 70)
print("YEARLY PEARSON CORRELATIONS WITH CHLOROPHYLL-a")
print("=" * 70)

print(
    temporal_corr_df.round(4)
)


# ------------------------------------------------------------
# 4. Add absolute correlation
# ------------------------------------------------------------

temporal_corr_df["absolute_r"] = (
    temporal_corr_df["pearson_r"].abs()
)


print("\n" + "=" * 70)
print("CORRELATION STRENGTH SUMMARY")
print("=" * 70)

print(
    temporal_corr_df.round(4)
)


# ------------------------------------------------------------
# 5. Save correlation results
# ------------------------------------------------------------

eda_folder = (
    "/content/drive/MyDrive/"
    "Indhi_Marine_Research_Data/03_eda"
)

os.makedirs(
    eda_folder,
    exist_ok=True
)


corr_path = os.path.join(
    eda_folder,
    "yearly_environmental_chlorophyll_correlations.csv"
)

temporal_corr_df.to_csv(
    corr_path
)

print(
    "\nCorrelation results saved to:"
)

print(corr_path)


# ------------------------------------------------------------
# 6. Scatter plots
# ------------------------------------------------------------

for variable in environmental_variables:

    plt.figure(figsize=(8, 5))

    plt.scatter(
        yearly_data[variable],
        yearly_data[target],
        alpha=0.75
    )

    plt.xlabel(variable)
    plt.ylabel("Mean Chlorophyll-a (mg/m³)")

    plt.title(
        f"Yearly Mean {variable} vs Chlorophyll-a"
    )

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()

    figure_path = os.path.join(
        eda_folder,
        f"yearly_{variable}_vs_chlorophyll_a.png"
    )

    plt.savefig(
        figure_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    print(
        f"Figure saved: {figure_path}"
    )


# ------------------------------------------------------------
# 7. Combined temporal relationship table
# ------------------------------------------------------------

relationship_table = yearly_data[
    [
        "year",
        "sst",
        "sss",
        "nitrate",
        "phosphate",
        "silicate",
        "chlorophyll_a"
    ]
].copy()


relationship_path = os.path.join(
    eda_folder,
    "yearly_environmental_chlorophyll_relationships.csv"
)

relationship_table.to_csv(
    relationship_path,
    index=False
)


print("\n" + "=" * 70)
print("STEP 31.9I COMPLETE")
print("=" * 70)

print(
    "Correlation file:"
)

print(corr_path)

print(
    "\nRelationship table:"
)

print(relationship_path)
```

```python
# ============================================================
# STEP 31.9L — LOCATION-WISE DAILY LAG / CROSS-CORRELATION
# ============================================================

import pandas as pd
import numpy as np
import os

print("=" * 70)
print("STEP 31.9L — DAILY LAG / CROSS-CORRELATION ANALYSIS")
print("=" * 70)

# ============================================================
# 1. LOAD FINAL DATASET
# ============================================================

DATA_PATH = (
    "/content/drive/MyDrive/Indhi_Marine_Research_Data/"
    "Western_Southern_Chlorophyll_Productivity_Final.csv"
)

df = pd.read_csv(
    DATA_PATH,
    parse_dates=["date"]
)

print("\nDataset shape:")
print(df.shape)

print("\nDate range:")
print(df["date"].min(), "to", df["date"].max())

print("\nSpatial locations:")
print(
    df[["latitude", "longitude"]]
    .drop_duplicates()
    .shape[0]
)

# ============================================================
# 2. VARIABLES
# ============================================================

environmental_variables = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate"
]

target = "chlorophyll_a"

lags = [1, 2, 3, 5, 7, 14, 30]

# ============================================================
# 3. SORT DATA
# ============================================================

df = df.sort_values(
    ["latitude", "longitude", "date"]
).reset_index(drop=True)

# ============================================================
# 4. CREATE LOCATION IDENTIFIER
# ============================================================

df["location"] = (
    df["latitude"].astype(str)
    + "_"
    + df["longitude"].astype(str)
)

print("\nNumber of locations:")
print(df["location"].nunique())

# ============================================================
# 5. CHECK UNIQUE DATES
# ============================================================

unique_dates = pd.DatetimeIndex(
    df["date"].drop_duplicates().sort_values()
)

print("\nNumber of dates in dataset:")
print(len(unique_dates))

# ============================================================
# 6. CREATE COMPLETE DAILY DATE RANGE
# ============================================================

full_dates = pd.date_range(
    start=df["date"].min(),
    end=df["date"].max(),
    freq="D"
)

print("\nExpected calendar dates:")
print(len(full_dates))

print("\nMissing calendar dates:")
print(
    len(full_dates.difference(unique_dates))
)

# ============================================================
# 7. GET UNIQUE SPATIAL LOCATIONS
# ============================================================

locations = (
    df[
        ["latitude", "longitude", "location"]
    ]
    .drop_duplicates()
    .sort_values(["latitude", "longitude"])
    .reset_index(drop=True)
)

print("\nNumber of spatial locations:")
print(len(locations))

# ============================================================
# 8. CREATE COMPLETE DATE × LOCATION INDEX
# ============================================================

complete_index = pd.MultiIndex.from_product(
    [
        full_dates,
        locations["location"]
    ],
    names=["date", "location"]
)

# ============================================================
# 9. PREPARE LOCATION INFORMATION
# ============================================================

location_info = locations.set_index("location")

# ============================================================
# 10. REINDEX DATASET
# ============================================================

df_lag = (
    df[
        [
            "date",
            "location",
            "sst",
            "sss",
            "nitrate",
            "phosphate",
            "silicate",
            "chlorophyll_a"
        ]
    ]
    .set_index(["date", "location"])
    .reindex(complete_index)
)

# ============================================================
# 11. RESTORE LATITUDE / LONGITUDE
# ============================================================

df_lag = df_lag.reset_index()

df_lag["latitude"] = df_lag["location"].map(
    location_info["latitude"]
)

df_lag["longitude"] = df_lag["location"].map(
    location_info["longitude"]
)

# ============================================================
# 12. SORT
# ============================================================

df_lag = df_lag.sort_values(
    ["location", "date"]
).reset_index(drop=True)

print("\nComplete daily grid shape:")
print(df_lag.shape)

print("\nMissing values after calendar completion:")
print(
    df_lag[
        environmental_variables + [target]
    ].isna().sum()
)

# ============================================================
# 13. CALCULATE LOCATION-WISE LAG CORRELATIONS
# ============================================================

results = []

print("\n")
print("=" * 70)
print("CALCULATING LOCATION-WISE LAG CORRELATIONS")
print("=" * 70)

for variable in environmental_variables:

    for lag in lags:

        lag_column = f"{variable}_lag_{lag}d"

        # Previous environmental value
        df_lag[lag_column] = (
            df_lag
            .groupby("location")[variable]
            .shift(lag)
        )

        # ----------------------------------------------------
        # Calculate Pearson correlation separately
        # for each spatial location
        # ----------------------------------------------------

        location_correlations = []

        for location, group in df_lag.groupby("location"):

            valid = group[
                [lag_column, target]
            ].dropna()

            if len(valid) >= 30:

                r = valid[lag_column].corr(
                    valid[target]
                )

                if pd.notna(r):
                    location_correlations.append(r)

        location_correlations = np.array(
            location_correlations,
            dtype=float
        )

        # ----------------------------------------------------
        # Summary across 102 locations
        # ----------------------------------------------------

        results.append({

            "variable": variable,

            "lag_days": lag,

            "number_of_locations": len(
                location_correlations
            ),

            "mean_r": np.mean(
                location_correlations
            ),

            "median_r": np.median(
                location_correlations
            ),

            "std_r": np.std(
                location_correlations,
                ddof=1
            ),

            "min_r": np.min(
                location_correlations
            ),

            "max_r": np.max(
                location_correlations
            ),

            "mean_absolute_r": np.mean(
                np.abs(location_correlations)
            )

        })

        print(
            f"{variable:10s} | "
            f"Lag {lag:2d} days | "
            f"Mean r = "
            f"{np.mean(location_correlations): .4f}"
        )

# ============================================================
# 14. CREATE RESULTS DATAFRAME
# ============================================================

lag_results = pd.DataFrame(results)

# ============================================================
# 15. DISPLAY RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("LOCATION-WISE DAILY LAG CORRELATION RESULTS")
print("=" * 70)

print(
    lag_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ============================================================
# 16. SAVE RESULTS
# ============================================================

EDA_DIR = (
    "/content/drive/MyDrive/Indhi_Marine_Research_Data/"
    "03_eda"
)

os.makedirs(
    EDA_DIR,
    exist_ok=True
)

lag_results_path = os.path.join(
    EDA_DIR,
    "daily_location_wise_lag_correlations.csv"
)

lag_results.to_csv(
    lag_results_path,
    index=False
)

print("\nResults saved to:")
print(lag_results_path)

print("\n")
print("=" * 70)
print("STEP 31.9L COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 31.9O — CHL EXTREME-EVENT PERSISTENCE ANALYSIS
# ======================================================================

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/Indhi_Marine_Research_Data")
EDA_DIR = BASE_DIR / "03_eda"
EDA_DIR.mkdir(parents=True, exist_ok=True)

# ----------------------------------------------------------------------
# LOAD FINAL DATASET
# ----------------------------------------------------------------------

# Use the final combined dataset already created in the project.
# Update this path ONLY if your final CSV has a different filename.

# Reuse the final dataset already loaded in memory
df = df.copy()

df["date"] = pd.to_datetime(df["date"])

print("=" * 70)
print("STEP 31.9O — CHL EXTREME-EVENT PERSISTENCE ANALYSIS")
print("=" * 70)

print("\nDataset shape:")
print(df.shape)

print("\nDate range:")
print(df["date"].min(), "to", df["date"].max())

print("\nNumber of spatial locations:")
print(df[["latitude", "longitude"]].drop_duplicates().shape[0])

# ----------------------------------------------------------------------
# 99th PERCENTILE THRESHOLD
# ----------------------------------------------------------------------

CHL_THRESHOLD = df["chlorophyll_a"].quantile(0.99)

print("\n" + "=" * 70)
print("EXTREME CHL THRESHOLD")
print("=" * 70)

print(f"99th percentile threshold: {CHL_THRESHOLD:.6f} mg/m³")

# ----------------------------------------------------------------------
# CREATE COMPLETE DAILY GRID
# ----------------------------------------------------------------------
# This ensures that missing calendar dates are treated as gaps.
# An extreme event must occur on genuinely consecutive calendar days.

locations = (
    df[["latitude", "longitude"]]
    .drop_duplicates()
    .sort_values(["latitude", "longitude"])
    .reset_index(drop=True)
)

all_dates = pd.date_range(
    start=df["date"].min(),
    end=df["date"].max(),
    freq="D"
)

print("\nExpected calendar dates:", len(all_dates))
print("Actual dates:", df["date"].nunique())
print("Missing calendar dates:", len(all_dates) - df["date"].nunique())

# ----------------------------------------------------------------------
# IDENTIFY EXTREME OBSERVATIONS
# ----------------------------------------------------------------------

extreme = df[df["chlorophyll_a"] >= CHL_THRESHOLD].copy()

extreme = extreme.sort_values(
    ["latitude", "longitude", "date"]
).reset_index(drop=True)

print("\n" + "=" * 70)
print("EXTREME OBSERVATIONS")
print("=" * 70)

print("Number of extreme observations:", len(extreme))

print(
    "Percentage of dataset:",
    round(len(extreme) / len(df) * 100, 4),
    "%"
)

# ----------------------------------------------------------------------
# IDENTIFY CONSECUTIVE EXTREME EVENTS
# ----------------------------------------------------------------------
# An event is defined separately for each spatial location.
#
# Consecutive dates:
#     2020-09-08
#     2020-09-09
#     2020-09-10
#
# belong to the same event.
#
# If there is a gap:
#     2020-09-08
#     2020-09-10
#
# they are treated as separate events.

extreme["previous_date"] = (
    extreme.groupby(["latitude", "longitude"])["date"]
    .shift(1)
)

extreme["date_difference"] = (
    extreme["date"] - extreme["previous_date"]
).dt.days

# New event whenever:
# 1. there is no previous extreme observation at that location
# OR
# 2. the previous extreme observation is not exactly one day earlier

extreme["new_event"] = (
    extreme["previous_date"].isna()
    | (extreme["date_difference"] != 1)
)

extreme["event_number"] = (
    extreme.groupby(["latitude", "longitude"])["new_event"]
    .cumsum()
)

# Create a unique event ID
extreme["event_id"] = (
    extreme["latitude"].round(6).astype(str)
    + "_"
    + extreme["longitude"].round(6).astype(str)
    + "_E"
    + extreme["event_number"].astype(str)
)

# ----------------------------------------------------------------------
# SUMMARISE EVENTS
# ----------------------------------------------------------------------

events = (
    extreme
    .groupby(
        [
            "event_id",
            "latitude",
            "longitude",
            "event_number"
        ]
    )
    .agg(
        start_date=("date", "min"),
        end_date=("date", "max"),
        duration_days=("date", "nunique"),
        observation_count=("chlorophyll_a", "count"),
        mean_chl=("chlorophyll_a", "mean"),
        median_chl=("chlorophyll_a", "median"),
        max_chl=("chlorophyll_a", "max"),
        min_chl=("chlorophyll_a", "min"),
        std_chl=("chlorophyll_a", "std"),
    )
    .reset_index()
)

events["std_chl"] = events["std_chl"].fillna(0)

# Calendar duration check
events["calendar_duration_days"] = (
    events["end_date"] - events["start_date"]
).dt.days + 1

# Year and month of event start
events["start_year"] = events["start_date"].dt.year
events["start_month"] = events["start_date"].dt.month
events["start_month_name"] = events["start_date"].dt.month_name()

# ----------------------------------------------------------------------
# CLASSIFY EVENT PERSISTENCE
# ----------------------------------------------------------------------

def classify_event(duration):
    if duration == 1:
        return "1-day isolated"
    elif duration == 2:
        return "2-day event"
    elif duration == 3:
        return "3-day event"
    elif duration <= 7:
        return "4-7 day event"
    elif duration <= 14:
        return "8-14 day event"
    else:
        return "15+ day event"

events["persistence_category"] = (
    events["duration_days"].apply(classify_event)
)

# ----------------------------------------------------------------------
# BASIC EVENT SUMMARY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("EXTREME CHLOROPHYLL-a EVENTS")
print("=" * 70)

print("Total extreme events:", len(events))

print(
    "1-day events:",
    (events["duration_days"] == 1).sum()
)

print(
    "Multi-day events:",
    (events["duration_days"] >= 2).sum()
)

print(
    "Events lasting 3+ days:",
    (events["duration_days"] >= 3).sum()
)

print(
    "Events lasting 7+ days:",
    (events["duration_days"] >= 7).sum()
)

print(
    "Longest event:",
    events["duration_days"].max(),
    "days"
)

# ----------------------------------------------------------------------
# PERSISTENCE DISTRIBUTION
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("EVENT PERSISTENCE DISTRIBUTION")
print("=" * 70)

persistence_summary = (
    events["persistence_category"]
    .value_counts()
    .reindex(
        [
            "1-day isolated",
            "2-day event",
            "3-day event",
            "4-7 day event",
            "8-14 day event",
            "15+ day event"
        ],
        fill_value=0
    )
    .rename_axis("persistence_category")
    .reset_index(name="event_count")
)

persistence_summary["percentage"] = (
    persistence_summary["event_count"]
    / len(events)
    * 100
)

print(persistence_summary.to_string(index=False))

# ----------------------------------------------------------------------
# LONGEST EVENTS
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP 20 LONGEST EXTREME CHL EVENTS")
print("=" * 70)

top_longest = (
    events
    .sort_values(
        ["duration_days", "max_chl"],
        ascending=[False, False]
    )
    .head(20)
)

print(
    top_longest[
        [
            "event_id",
            "latitude",
            "longitude",
            "start_date",
            "end_date",
            "duration_days",
            "mean_chl",
            "max_chl"
        ]
    ].to_string(index=False)
)

# ----------------------------------------------------------------------
# STRONGEST EVENTS BY MAXIMUM CHL
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP 20 STRONGEST EXTREME CHL EVENTS")
print("=" * 70)

top_strongest = (
    events
    .sort_values(
        ["max_chl", "duration_days"],
        ascending=[False, False]
    )
    .head(20)
)

print(
    top_strongest[
        [
            "event_id",
            "latitude",
            "longitude",
            "start_date",
            "end_date",
            "duration_days",
            "mean_chl",
            "max_chl"
        ]
    ].to_string(index=False)
)

# ----------------------------------------------------------------------
# MULTI-DAY EVENTS ONLY
# ----------------------------------------------------------------------

multi_day_events = events[
    events["duration_days"] >= 2
].copy()

print("\n" + "=" * 70)
print("MULTI-DAY EXTREME EVENTS")
print("=" * 70)

print("Number of multi-day events:", len(multi_day_events))

if len(multi_day_events) > 0:

    print(
        "\nLongest multi-day event:",
        multi_day_events["duration_days"].max(),
        "days"
    )

    print(
        "Mean multi-day event duration:",
        round(multi_day_events["duration_days"].mean(), 2),
        "days"
    )

# ----------------------------------------------------------------------
# EVENTS BY YEAR
# ----------------------------------------------------------------------

events_by_year = (
    events
    .groupby("start_year")
    .agg(
        total_events=("event_id", "count"),
        multi_day_events=(
            "duration_days",
            lambda x: (x >= 2).sum()
        ),
        events_3plus_days=(
            "duration_days",
            lambda x: (x >= 3).sum()
        ),
        longest_event_days=("duration_days", "max"),
        maximum_chl=("max_chl", "max"),
        mean_event_chl=("mean_chl", "mean")
    )
    .reset_index()
)

print("\n" + "=" * 70)
print("EXTREME EVENTS BY YEAR")
print("=" * 70)

print(events_by_year.to_string(index=False))

# ----------------------------------------------------------------------
# EVENTS BY MONTH
# ----------------------------------------------------------------------

events_by_month = (
    events
    .groupby("start_month")
    .agg(
        total_events=("event_id", "count"),
        multi_day_events=(
            "duration_days",
            lambda x: (x >= 2).sum()
        ),
        longest_event_days=("duration_days", "max"),
        maximum_chl=("max_chl", "max"),
        mean_event_chl=("mean_chl", "mean")
    )
    .reset_index()
)

events_by_month["month_name"] = (
    pd.to_datetime(
        events_by_month["start_month"],
        format="%m"
    ).dt.month_name()
)

print("\n" + "=" * 70)
print("EXTREME EVENTS BY MONTH")
print("=" * 70)

print(events_by_month.to_string(index=False))

# ----------------------------------------------------------------------
# EVENTS BY LOCATION
# ----------------------------------------------------------------------

events_by_location = (
    events
    .groupby(["latitude", "longitude"])
    .agg(
        total_events=("event_id", "count"),
        multi_day_events=(
            "duration_days",
            lambda x: (x >= 2).sum()
        ),
        longest_event_days=("duration_days", "max"),
        maximum_chl=("max_chl", "max"),
        mean_event_chl=("mean_chl", "mean")
    )
    .reset_index()
)

print("\n" + "=" * 70)
print("TOP LOCATIONS BY NUMBER OF EXTREME EVENTS")
print("=" * 70)

print(
    events_by_location
    .sort_values("total_events", ascending=False)
    .head(20)
    .to_string(index=False)
)

# ----------------------------------------------------------------------
# SAVE EVENT DATA
# ----------------------------------------------------------------------

events_file = EDA_DIR / "chlorophyll_extreme_events.csv"
persistence_file = EDA_DIR / "chlorophyll_extreme_event_persistence.csv"
year_file = EDA_DIR / "chlorophyll_extreme_events_by_year.csv"
month_file = EDA_DIR / "chlorophyll_extreme_events_by_month.csv"
location_file = EDA_DIR / "chlorophyll_extreme_events_by_location.csv"

events.to_csv(events_file, index=False)
persistence_summary.to_csv(persistence_file, index=False)
events_by_year.to_csv(year_file, index=False)
events_by_month.to_csv(month_file, index=False)
events_by_location.to_csv(location_file, index=False)

# ----------------------------------------------------------------------
# FIGURE 1 — EVENT DURATION DISTRIBUTION
# ----------------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(
    events["duration_days"],
    bins=np.arange(
        0.5,
        events["duration_days"].max() + 1.5,
        1
    )
)

plt.xlabel("Extreme Event Duration (days)")
plt.ylabel("Number of Events")
plt.title(
    "Persistence of Extreme Chlorophyll-a Events\n"
    "Western and Southern Sri Lankan Coastal Waters"
)

plt.grid(alpha=0.3)
plt.tight_layout()

duration_plot = EDA_DIR / "chlorophyll_extreme_event_duration.png"
plt.savefig(duration_plot, dpi=300, bbox_inches="tight")
plt.show()

# ----------------------------------------------------------------------
# FIGURE 2 — PERSISTENCE CATEGORY DISTRIBUTION
# ----------------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    persistence_summary["persistence_category"],
    persistence_summary["event_count"]
)

plt.xlabel("Event Persistence Category")
plt.ylabel("Number of Events")
plt.title(
    "Persistence Categories of Extreme Chlorophyll-a Events"
)

plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

category_plot = EDA_DIR / "chlorophyll_extreme_event_persistence_categories.png"
plt.savefig(category_plot, dpi=300, bbox_inches="tight")
plt.show()

# ----------------------------------------------------------------------
# FIGURE 3 — MULTI-DAY EXTREME EVENTS BY YEAR
# ----------------------------------------------------------------------

plt.figure(figsize=(11, 6))

plt.plot(
    events_by_year["start_year"],
    events_by_year["multi_day_events"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Number of Multi-day Extreme Events")
plt.title(
    "Multi-day Extreme Chlorophyll-a Events by Year"
)

plt.grid(alpha=0.3)
plt.tight_layout()

year_plot = EDA_DIR / "chlorophyll_multiday_extreme_events_by_year.png"
plt.savefig(year_plot, dpi=300, bbox_inches="tight")
plt.show()

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 31.9O COMPLETE")
print("=" * 70)

print(f"99th percentile threshold: {CHL_THRESHOLD:.6f} mg/m³")
print(f"Total extreme observations: {len(extreme):,}")
print(f"Total extreme events: {len(events):,}")
print(
    f"Multi-day events: "
    f"{(events['duration_days'] >= 2).sum():,}"
)
print(
    f"Events lasting 3+ days: "
    f"{(events['duration_days'] >= 3).sum():,}"
)
print(
    f"Longest event: "
    f"{events['duration_days'].max()} days"
)

print("\nFiles saved:")
print(events_file)
print(persistence_file)
print(year_file)
print(month_file)
print(location_file)
print(duration_plot)
print(category_plot)
print(year_plot)
```

```python
# ======================================================================
# STEP 31.9P-1 — CLEAN DAILY FEATURE-ENGINEERING BASE
# ======================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 31.9P-1 — CLEAN DAILY FEATURE-ENGINEERING BASE")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/Indhi_Marine_Research_Data")

DATA_FILE = BASE_DIR / "Western_Southern_Chlorophyll_Productivity_Final.csv"

EDA_DIR = BASE_DIR / "03_eda"
FEATURE_DIR = BASE_DIR / "04_feature_engineering"

FEATURE_DIR.mkdir(parents=True, exist_ok=True)

print("\nData file:")
print(DATA_FILE)

# ----------------------------------------------------------------------
# 2. LOAD FINAL DATASET
# ----------------------------------------------------------------------

df = pd.read_csv(DATA_FILE)

print("\n" + "=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

# ----------------------------------------------------------------------
# 3. CONVERT DATE
# ----------------------------------------------------------------------

df["date"] = pd.to_datetime(df["date"])

# ----------------------------------------------------------------------
# 4. SORT BY LOCATION AND DATE
# ----------------------------------------------------------------------

df = df.sort_values(
    ["latitude", "longitude", "date"]
).reset_index(drop=True)

# ----------------------------------------------------------------------
# 5. BASIC INFORMATION
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("BASIC DATASET INFORMATION")
print("=" * 70)

print("Rows:", len(df))

print("Date range:")
print(df["date"].min(), "to", df["date"].max())

print("Number of spatial locations:",
      df[["latitude", "longitude"]].drop_duplicates().shape[0])

# ----------------------------------------------------------------------
# 6. CHECK DUPLICATE LOCATION-DATE RECORDS
# ----------------------------------------------------------------------

duplicate_mask = df.duplicated(
    subset=["latitude", "longitude", "date"],
    keep=False
)

duplicate_count = duplicate_mask.sum()

print("\n" + "=" * 70)
print("DUPLICATE LOCATION-DATE CHECK")
print("=" * 70)

print("Duplicate rows:", duplicate_count)

if duplicate_count == 0:
    print("✓ No duplicate location-date records found.")
else:
    print("⚠ Duplicate location-date records detected.")

# ----------------------------------------------------------------------
# 7. CHECK MISSING VALUES
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUE CHECK")
print("=" * 70)

missing_values = df[
    [
        "sst",
        "sss",
        "nitrate",
        "phosphate",
        "silicate",
        "chlorophyll_a"
    ]
].isna().sum()

print(missing_values)

# ----------------------------------------------------------------------
# 8. CHECK DATE CONTINUITY FOR EACH LOCATION
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("LOCATION-WISE DATE CONTINUITY CHECK")
print("=" * 70)

expected_start = df["date"].min()
expected_end = df["date"].max()

expected_dates = pd.date_range(
    start=expected_start,
    end=expected_end,
    freq="D"
)

expected_days = len(expected_dates)

print("Expected calendar days:", expected_days)

location_summary = []

for (lat, lon), group in df.groupby(
    ["latitude", "longitude"]
):

    actual_dates = pd.DatetimeIndex(
        group["date"].drop_duplicates()
    )

    missing_dates = expected_dates.difference(actual_dates)

    location_summary.append({
        "latitude": lat,
        "longitude": lon,
        "actual_days": len(actual_dates),
        "expected_days": expected_days,
        "missing_days": len(missing_dates),
        "first_date": actual_dates.min(),
        "last_date": actual_dates.max()
    })

location_summary_df = pd.DataFrame(location_summary)

print("\nLocation continuity summary:")
print(location_summary_df.head())

print("\nLocations with missing dates:",
      (location_summary_df["missing_days"] > 0).sum())

print("Total missing location-days:",
      location_summary_df["missing_days"].sum())

print("Maximum missing days at one location:",
      location_summary_df["missing_days"].max())

# ----------------------------------------------------------------------
# 9. IDENTIFY THE GLOBAL MISSING CALENDAR DATES
# ----------------------------------------------------------------------

actual_global_dates = pd.DatetimeIndex(
    df["date"].drop_duplicates()
)

missing_global_dates = expected_dates.difference(
    actual_global_dates
)

print("\n" + "=" * 70)
print("GLOBAL CALENDAR CHECK")
print("=" * 70)

print("Expected calendar dates:", expected_days)
print("Actual calendar dates:", len(actual_global_dates))
print("Missing calendar dates:", len(missing_global_dates))

if len(missing_global_dates) > 0:
    print("\nMissing dates:")
    print(missing_global_dates.tolist())

# ----------------------------------------------------------------------
# 10. VERIFY EACH LOCATION HAS THE SAME NUMBER OF OBSERVATIONS
# ----------------------------------------------------------------------

location_counts = (
    df.groupby(["latitude", "longitude"])
      .size()
      .reset_index(name="observation_count")
)

print("\n" + "=" * 70)
print("OBSERVATIONS PER LOCATION")
print("=" * 70)

print(location_counts["observation_count"].describe())

print(
    "\nLocations with the most observations:",
    location_counts["observation_count"].max()
)

print(
    "Locations with the fewest observations:",
    location_counts["observation_count"].min()
)

# ----------------------------------------------------------------------
# 11. CHECK CHRONOLOGICAL ORDER
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("CHRONOLOGICAL ORDER CHECK")
print("=" * 70)

chronological_check = (
    df.groupby(["latitude", "longitude"])["date"]
      .apply(lambda x: x.is_monotonic_increasing)
)

print(
    "Locations with correctly ordered dates:",
    chronological_check.sum(),
    "/",
    len(chronological_check)
)

# ----------------------------------------------------------------------
# 12. SAVE CLEAN BASE DATASET
# ----------------------------------------------------------------------

CLEAN_FILE = FEATURE_DIR / "daily_feature_engineering_base.csv"

df.to_csv(
    CLEAN_FILE,
    index=False
)

# Save location continuity information
LOCATION_CHECK_FILE = FEATURE_DIR / "location_date_continuity_check.csv"

location_summary_df.to_csv(
    LOCATION_CHECK_FILE,
    index=False
)

# ----------------------------------------------------------------------
# 13. FINAL SUMMARY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 31.9P-1 COMPLETE")
print("=" * 70)

print("Clean base shape:", df.shape)

print("Number of locations:",
      df[["latitude", "longitude"]].drop_duplicates().shape[0])

print("Date range:",
      df["date"].min(),
      "to",
      df["date"].max())

print("\nClean base saved to:")
print(CLEAN_FILE)

print("\nLocation continuity report saved to:")
print(LOCATION_CHECK_FILE)

print("\nNext step:")
print("STEP 31.9P-2 — TEMPORAL FEATURE ENGINEERING")
```

```python
# ======================================================================
# STEP 31.9P-2 — TEMPORAL FEATURE ENGINEERING
# ======================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 31.9P-2 — TEMPORAL FEATURE ENGINEERING")
print("=" * 70)

# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/Indhi_Marine_Research_Data")

INPUT_FILE = (
    BASE_DIR
    / "04_feature_engineering"
    / "daily_feature_engineering_base.csv"
)

OUTPUT_DIR = BASE_DIR / "04_feature_engineering"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = (
    OUTPUT_DIR
    / "daily_temporal_features.csv"
)

# ----------------------------------------------------------------------
# LOAD CLEAN BASE
# ----------------------------------------------------------------------

df = pd.read_csv(INPUT_FILE)

df["date"] = pd.to_datetime(df["date"])

print("\n" + "=" * 70)
print("DATASET LOADED")
print("=" * 70)

print(f"Shape: {df.shape}")
print(f"Date range: {df['date'].min()} to {df['date'].max()}")
print(f"Number of locations: {df[['latitude', 'longitude']].drop_duplicates().shape[0]}")

# ----------------------------------------------------------------------
# TEMPORAL FEATURES
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("CREATING TEMPORAL FEATURES")
print("=" * 70)

# Basic calendar features
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["day_of_year"] = df["date"].dt.dayofyear
df["day_of_week"] = df["date"].dt.dayofweek
df["week_of_year"] = df["date"].dt.isocalendar().week.astype(int)

# ----------------------------------------------------------------------
# SEASON
# ----------------------------------------------------------------------
# Sri Lankan monsoon-based seasons used in the earlier EDA:
#
# Northeast Monsoon      : December - February
# First Inter-monsoon    : March - April
# Southwest Monsoon      : May - September
# Second Inter-monsoon   : October - November
# ----------------------------------------------------------------------

def assign_season(month):

    if month in [12, 1, 2]:
        return "Northeast Monsoon"

    elif month in [3, 4]:
        return "First Inter-monsoon"

    elif month in [5, 6, 7, 8, 9]:
        return "Southwest Monsoon"

    elif month in [10, 11]:
        return "Second Inter-monsoon"

    return "Unknown"


df["season"] = df["month"].apply(assign_season)

# ----------------------------------------------------------------------
# CYCLIC DAY-OF-YEAR FEATURES
# ----------------------------------------------------------------------
# These represent the annual cycle continuously.
#
# Example:
# December and January are close in the cycle rather than being
# treated as far apart numbers.
# ----------------------------------------------------------------------

days_in_year = np.where(
    df["date"].dt.is_leap_year,
    366,
    365
)

df["day_of_year_sin"] = np.sin(
    2 * np.pi * (df["day_of_year"] - 1) / days_in_year
)

df["day_of_year_cos"] = np.cos(
    2 * np.pi * (df["day_of_year"] - 1) / days_in_year
)

# ----------------------------------------------------------------------
# CYCLIC MONTH FEATURES
# ----------------------------------------------------------------------

df["month_sin"] = np.sin(
    2 * np.pi * (df["month"] - 1) / 12
)

df["month_cos"] = np.cos(
    2 * np.pi * (df["month"] - 1) / 12
)

# ----------------------------------------------------------------------
# CYCLIC DAY-OF-WEEK FEATURES
# ----------------------------------------------------------------------

df["day_of_week_sin"] = np.sin(
    2 * np.pi * df["day_of_week"] / 7
)

df["day_of_week_cos"] = np.cos(
    2 * np.pi * df["day_of_week"] / 7
)

# ----------------------------------------------------------------------
# CHECK RESULTS
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TEMPORAL FEATURES CREATED")
print("=" * 70)

temporal_columns = [
    "year",
    "month",
    "day",
    "day_of_year",
    "day_of_week",
    "week_of_year",
    "season",
    "day_of_year_sin",
    "day_of_year_cos",
    "month_sin",
    "month_cos",
    "day_of_week_sin",
    "day_of_week_cos"
]

print("\nNew temporal columns:")
for column in temporal_columns:
    print(f"✓ {column}")

# ----------------------------------------------------------------------
# SEASON DISTRIBUTION
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("SEASON DISTRIBUTION")
print("=" * 70)

season_counts = df["season"].value_counts()

print(season_counts)

# ----------------------------------------------------------------------
# TEMPORAL FEATURE RANGE CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TEMPORAL FEATURE CHECK")
print("=" * 70)

print(
    df[
        [
            "date",
            "year",
            "month",
            "day",
            "day_of_year",
            "day_of_week",
            "week_of_year",
            "season"
        ]
    ].head(10)
)

# ----------------------------------------------------------------------
# MISSING VALUE CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUE CHECK")
print("=" * 70)

print(df[temporal_columns].isna().sum())

# ----------------------------------------------------------------------
# FINAL STRUCTURE
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL DATASET STRUCTURE")
print("=" * 70)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")

print("\nColumns:")
print(df.columns.tolist())

# ----------------------------------------------------------------------
# SAVE
# ----------------------------------------------------------------------

df.to_csv(OUTPUT_FILE, index=False)

print("\n" + "=" * 70)
print("STEP 31.9P-2 COMPLETE")
print("=" * 70)

print(f"Temporal feature dataset saved to:")
print(OUTPUT_FILE)
```

```python
# ======================================================================
# STEP 31.9P-5 — PAST-ONLY ROLLING FEATURES
# ======================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 31.9P-5 — PAST-ONLY ROLLING FEATURES")
print("=" * 70)


# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

INPUT_FILE = (
    BASE_DIR
    / "04_feature_engineering"
    / "daily_environmental_lag_features.csv"
)

OUTPUT_DIR = BASE_DIR / "04_feature_engineering"

OUTPUT_FILE = (
    OUTPUT_DIR
    / "daily_rolling_features.csv"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ----------------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------------

df = pd.read_csv(
    INPUT_FILE,
    parse_dates=["date"]
)

print("\n" + "=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Shape:", df.shape)

print(
    "Date range:",
    df["date"].min(),
    "to",
    df["date"].max()
)

print(
    "Number of spatial locations:",
    df[
        ["latitude", "longitude"]
    ].drop_duplicates().shape[0]
)


# ----------------------------------------------------------------------
# SORT CHRONOLOGICALLY
# ----------------------------------------------------------------------

df = df.sort_values(
    ["latitude", "longitude", "date"]
).reset_index(drop=True)


# ----------------------------------------------------------------------
# VARIABLES
# ----------------------------------------------------------------------

CHL_VARIABLE = "chlorophyll_a"

ENVIRONMENTAL_VARIABLES = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate"
]

print("\n" + "=" * 70)
print("ROLLING FEATURE CONFIGURATION")
print("=" * 70)

print("CHL variable:", CHL_VARIABLE)

print("Environmental variables:")
for variable in ENVIRONMENTAL_VARIABLES:
    print("✓", variable)


# ----------------------------------------------------------------------
# CREATE LOCATION GROUP
# ----------------------------------------------------------------------

location_group = df.groupby(
    ["latitude", "longitude"],
    sort=False
)


# ======================================================================
# CHLOROPHYLL-a ROLLING FEATURES
# ======================================================================

print("\n" + "=" * 70)
print("CREATING CHLOROPHYLL-a ROLLING FEATURES")
print("=" * 70)


# ----------------------------------------------------------------------
# IMPORTANT:
#
# shift(1) means today's CHL is NOT included.
#
# For example:
#
# chl_rolling_mean_7 on 2020-08-20
#
# uses historical CHL observations before 2020-08-20,
# not the CHL value from 2020-08-20 itself.
# ----------------------------------------------------------------------

past_chl = location_group[CHL_VARIABLE].shift(1)


# 3-day rolling mean
df["chl_rolling_mean_3"] = (
    past_chl
    .groupby(
        [
            df["latitude"],
            df["longitude"]
        ]
    )
    .transform(
        lambda x: x.rolling(
            window=3,
            min_periods=3
        ).mean()
    )
)


# 7-day rolling mean
df["chl_rolling_mean_7"] = (
    past_chl
    .groupby(
        [
            df["latitude"],
            df["longitude"]
        ]
    )
    .transform(
        lambda x: x.rolling(
            window=7,
            min_periods=7
        ).mean()
    )
)


# 14-day rolling mean
df["chl_rolling_mean_14"] = (
    past_chl
    .groupby(
        [
            df["latitude"],
            df["longitude"]
        ]
    )
    .transform(
        lambda x: x.rolling(
            window=14,
            min_periods=14
        ).mean()
    )
)


# 7-day rolling standard deviation
df["chl_rolling_std_7"] = (
    past_chl
    .groupby(
        [
            df["latitude"],
            df["longitude"]
        ]
    )
    .transform(
        lambda x: x.rolling(
            window=7,
            min_periods=7
        ).std()
    )
)


# 14-day rolling standard deviation
df["chl_rolling_std_14"] = (
    past_chl
    .groupby(
        [
            df["latitude"],
            df["longitude"]
        ]
    )
    .transform(
        lambda x: x.rolling(
            window=14,
            min_periods=14
        ).std()
    )
)


print("✓ chl_rolling_mean_3")
print("✓ chl_rolling_mean_7")
print("✓ chl_rolling_mean_14")
print("✓ chl_rolling_std_7")
print("✓ chl_rolling_std_14")


# ======================================================================
# ENVIRONMENTAL ROLLING MEANS
# ======================================================================

print("\n" + "=" * 70)
print("CREATING ENVIRONMENTAL ROLLING FEATURES")
print("=" * 70)


ENVIRONMENTAL_WINDOWS = [3, 7, 14]

environmental_rolling_columns = []


for variable in ENVIRONMENTAL_VARIABLES:

    # --------------------------------------------------------------
    # Exclude today's environmental observation.
    # --------------------------------------------------------------

    past_values = location_group[variable].shift(1)

    for window in ENVIRONMENTAL_WINDOWS:

        column_name = (
            f"{variable}_rolling_mean_{window}"
        )

        df[column_name] = (
            past_values
            .groupby(
                [
                    df["latitude"],
                    df["longitude"]
                ]
            )
            .transform(
                lambda x, w=window:
                x.rolling(
                    window=w,
                    min_periods=w
                ).mean()
            )
        )

        environmental_rolling_columns.append(
            column_name
        )

        print(f"✓ {column_name}")


# ======================================================================
# ROLLING FEATURE SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("ROLLING FEATURE MISSING VALUE SUMMARY")
print("=" * 70)


rolling_columns = [
    "chl_rolling_mean_3",
    "chl_rolling_mean_7",
    "chl_rolling_mean_14",
    "chl_rolling_std_7",
    "chl_rolling_std_14"
] + environmental_rolling_columns


rolling_summary = []

for column in rolling_columns:

    valid_count = df[column].notna().sum()

    missing_count = df[column].isna().sum()

    missing_percentage = (
        df[column].isna().mean() * 100
    )

    rolling_summary.append({
        "feature": column,
        "valid_values": valid_count,
        "missing_values": missing_count,
        "missing_percentage": missing_percentage
    })


rolling_summary_df = pd.DataFrame(
    rolling_summary
)

print(
    rolling_summary_df.to_string(
        index=False
    )
)


# ======================================================================
# VERIFY ROLLING FEATURES
# ======================================================================

print("\n" + "=" * 70)
print("ROLLING FEATURE VERIFICATION")
print("=" * 70)


# Select first location
first_location = (
    df[
        ["latitude", "longitude"]
    ]
    .drop_duplicates()
    .iloc[0]
)

sample_lat = first_location["latitude"]
sample_lon = first_location["longitude"]


sample = df[
    (df["latitude"] == sample_lat) &
    (df["longitude"] == sample_lon)
].copy()


sample = sample.sort_values(
    "date"
)


verification_columns = [
    "date",
    "chlorophyll_a",
    "chl_lag_1",
    "chl_rolling_mean_3",
    "chl_rolling_mean_7",
    "chl_rolling_mean_14",
    "chl_rolling_std_7",
    "chl_rolling_std_14",
    "sst",
    "sst_rolling_mean_7",
    "phosphate",
    "phosphate_rolling_mean_7"
]


print(
    "\nSample location:"
    f" Latitude = {sample_lat},"
    f" Longitude = {sample_lon}"
)


print(
    sample[
        verification_columns
    ]
    .head(25)
    .to_string(index=False)
)


# ======================================================================
# CHECK ORIGINAL VARIABLES
# ======================================================================

print("\n" + "=" * 70)
print("ORIGINAL VARIABLE CHECK")
print("=" * 70)

original_variables = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate",
    "chlorophyll_a"
]

print(
    df[
        original_variables
    ].isna().sum()
)


# ======================================================================
# CHECK ROW COUNT
# ======================================================================

print("\n" + "=" * 70)
print("ROW COUNT CHECK")
print("=" * 70)

print("Rows:", len(df))


# ======================================================================
# FINAL STRUCTURE
# ======================================================================

print("\n" + "=" * 70)
print("FINAL DATASET STRUCTURE")
print("=" * 70)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nRolling features created:")

for column in rolling_columns:
    print("✓", column)


# ======================================================================
# SAVE ROLLING SUMMARY
# ======================================================================

ROLLING_SUMMARY_FILE = (
    OUTPUT_DIR
    / "rolling_feature_summary.csv"
)

rolling_summary_df.to_csv(
    ROLLING_SUMMARY_FILE,
    index=False
)


# ======================================================================
# SAVE DATASET
# ======================================================================

df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ======================================================================
# COMPLETE
# ======================================================================

print("\n" + "=" * 70)
print("STEP 31.9P-5 COMPLETE")
print("=" * 70)

print("\nFinal dataset shape:")
print(df.shape)

print("\nNumber of rolling features:")
print(len(rolling_columns))

print("\nDataset saved to:")
print(OUTPUT_FILE)

print("\nRolling feature summary saved to:")
print(ROLLING_SUMMARY_FILE)
```

```python
# ======================================================================
# STEP 31.9P-6 — SPATIAL FEATURES & CHL TRANSFORMATION
# ======================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("=" * 70)
print("STEP 31.9P-6 — SPATIAL FEATURES & CHL TRANSFORMATION")
print("=" * 70)


# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

INPUT_FILE = (
    BASE_DIR
    / "04_feature_engineering"
    / "daily_rolling_features.csv"
)

OUTPUT_DIR = BASE_DIR / "04_feature_engineering"

OUTPUT_FILE = (
    OUTPUT_DIR
    / "daily_spatial_transformed_features.csv"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ----------------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------------

df = pd.read_csv(
    INPUT_FILE,
    parse_dates=["date"]
)

print("\n" + "=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Shape:", df.shape)

print(
    "Date range:",
    df["date"].min(),
    "to",
    df["date"].max()
)

print(
    "Number of spatial locations:",
    df[
        ["latitude", "longitude"]
    ].drop_duplicates().shape[0]
)


# ======================================================================
# SPATIAL INFORMATION
# ======================================================================

print("\n" + "=" * 70)
print("SPATIAL INFORMATION")
print("=" * 70)

print(
    "Latitude range:",
    df["latitude"].min(),
    "to",
    df["latitude"].max()
)

print(
    "Longitude range:",
    df["longitude"].min(),
    "to",
    df["longitude"].max()
)

print(
    "Number of unique latitudes:",
    df["latitude"].nunique()
)

print(
    "Number of unique longitudes:",
    df["longitude"].nunique()
)

print(
    "Number of spatial locations:",
    df[
        ["latitude", "longitude"]
    ].drop_duplicates().shape[0]
)


# ======================================================================
# CREATE SPATIAL FEATURES
# ======================================================================

print("\n" + "=" * 70)
print("CREATING SPATIAL FEATURES")
print("=" * 70)


# ----------------------------------------------------------------------
# Keep the original coordinates.
#
# We create normalized versions so that models can use the spatial
# position without changing the original latitude/longitude values.
# ----------------------------------------------------------------------

lat_min = df["latitude"].min()
lat_max = df["latitude"].max()

lon_min = df["longitude"].min()
lon_max = df["longitude"].max()


df["latitude_normalized"] = (
    (df["latitude"] - lat_min)
    / (lat_max - lat_min)
)


df["longitude_normalized"] = (
    (df["longitude"] - lon_min)
    / (lon_max - lon_min)
)


print("✓ latitude_normalized")
print("✓ longitude_normalized")


# ----------------------------------------------------------------------
# Distance-like spatial coordinates from the centre of the study area.
#
# These are NOT physical distances in kilometres.
# They simply describe relative position around the study-area centre.
# ----------------------------------------------------------------------

latitude_center = df["latitude"].mean()
longitude_center = df["longitude"].mean()


df["latitude_centered"] = (
    df["latitude"] - latitude_center
)

df["longitude_centered"] = (
    df["longitude"] - longitude_center
)


print("✓ latitude_centered")
print("✓ longitude_centered")


# ----------------------------------------------------------------------
# Radial spatial position
#
# Relative distance from the centre in coordinate space.
# ----------------------------------------------------------------------

df["spatial_distance_from_center"] = np.sqrt(
    df["latitude_centered"] ** 2
    +
    df["longitude_centered"] ** 2
)

print("✓ spatial_distance_from_center")


# ======================================================================
# SPATIAL FEATURE CHECK
# ======================================================================

print("\n" + "=" * 70)
print("SPATIAL FEATURE CHECK")
print("=" * 70)

spatial_columns = [
    "latitude",
    "longitude",
    "latitude_normalized",
    "longitude_normalized",
    "latitude_centered",
    "longitude_centered",
    "spatial_distance_from_center"
]

print(
    df[spatial_columns].head(10).to_string(
        index=False
    )
)


# ======================================================================
# CHLOROPHYLL-a DISTRIBUTION
# ======================================================================

print("\n" + "=" * 70)
print("CHLOROPHYLL-a DISTRIBUTION")
print("=" * 70)

chl = df["chlorophyll_a"]

print(
    "Minimum:",
    chl.min()
)

print(
    "Maximum:",
    chl.max()
)

print(
    "Mean:",
    chl.mean()
)

print(
    "Median:",
    chl.median()
)

print(
    "Standard deviation:",
    chl.std()
)

print(
    "Skewness:",
    chl.skew()
)


# ======================================================================
# CHLOROPHYLL-a LOG TRANSFORMATION
# ======================================================================

print("\n" + "=" * 70)
print("CREATING log1p(CHL) TRANSFORMATION")
print("=" * 70)


# ----------------------------------------------------------------------
# log1p(x) = log(1 + x)
#
# This is appropriate for non-negative CHL values and reduces the
# influence of very large right-tail observations.
# ----------------------------------------------------------------------

df["chlorophyll_a_log1p"] = np.log1p(
    df["chlorophyll_a"]
)


print("✓ chlorophyll_a_log1p created")


# ======================================================================
# TRANSFORMED CHL STATISTICS
# ======================================================================

print("\n" + "=" * 70)
print("TRANSFORMED CHLOROPHYLL-a STATISTICS")
print("=" * 70)

chl_log = df["chlorophyll_a_log1p"]

print(
    "Minimum:",
    chl_log.min()
)

print(
    "Maximum:",
    chl_log.max()
)

print(
    "Mean:",
    chl_log.mean()
)

print(
    "Median:",
    chl_log.median()
)

print(
    "Standard deviation:",
    chl_log.std()
)

print(
    "Skewness:",
    chl_log.skew()
)


# ======================================================================
# ORIGINAL VS TRANSFORMED DISTRIBUTION
# ======================================================================

print("\n" + "=" * 70)
print("PLOTTING CHLOROPHYLL-a DISTRIBUTIONS")
print("=" * 70)


# ----------------------------------------------------------------------
# Original CHL
# ----------------------------------------------------------------------

plt.figure(figsize=(9, 5))

plt.hist(
    df["chlorophyll_a"],
    bins=100
)

plt.xlabel(
    "Chlorophyll-a (mg/m³)"
)

plt.ylabel(
    "Frequency"
)

plt.title(
    "Original Chlorophyll-a Distribution"
)

plt.tight_layout()

original_plot = (
    OUTPUT_DIR
    / "chlorophyll_a_original_distribution.png"
)

plt.savefig(
    original_plot,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "Original distribution saved to:"
)
print(original_plot)


# ----------------------------------------------------------------------
# Log-transformed CHL
# ----------------------------------------------------------------------

plt.figure(figsize=(9, 5))

plt.hist(
    df["chlorophyll_a_log1p"],
    bins=100
)

plt.xlabel(
    "log1p(Chlorophyll-a)"
)

plt.ylabel(
    "Frequency"
)

plt.title(
    "Log-Transformed Chlorophyll-a Distribution"
)

plt.tight_layout()

log_plot = (
    OUTPUT_DIR
    / "chlorophyll_a_log1p_distribution.png"
)

plt.savefig(
    log_plot,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "Log-transformed distribution saved to:"
)
print(log_plot)


# ======================================================================
# COMPARISON TABLE
# ======================================================================

print("\n" + "=" * 70)
print("ORIGINAL VS TRANSFORMED CHLOROPHYLL-a")
print("=" * 70)


comparison = pd.DataFrame({
    "measure": [
        "minimum",
        "maximum",
        "mean",
        "median",
        "standard_deviation",
        "skewness"
    ],
    "original_chlorophyll_a": [
        chl.min(),
        chl.max(),
        chl.mean(),
        chl.median(),
        chl.std(),
        chl.skew()
    ],
    "log1p_chlorophyll_a": [
        chl_log.min(),
        chl_log.max(),
        chl_log.mean(),
        chl_log.median(),
        chl_log.std(),
        chl_log.skew()
    ]
})


print(
    comparison.to_string(
        index=False
    )
)


# ======================================================================
# ORIGINAL TARGET INTEGRITY CHECK
# ======================================================================

print("\n" + "=" * 70)
print("ORIGINAL TARGET INTEGRITY CHECK")
print("=" * 70)


# Recalculate the original CHL statistics and verify that the original
# target has not been replaced by the transformed version.

print(
    "Original chlorophyll_a minimum:",
    df["chlorophyll_a"].min()
)

print(
    "Original chlorophyll_a maximum:",
    df["chlorophyll_a"].max()
)

print(
    "Original chlorophyll_a missing:",
    df["chlorophyll_a"].isna().sum()
)

print(
    "Transformed chlorophyll_a missing:",
    df["chlorophyll_a_log1p"].isna().sum()
)


# ======================================================================
# SPATIAL FEATURE MISSING VALUE CHECK
# ======================================================================

print("\n" + "=" * 70)
print("SPATIAL / TRANSFORMATION MISSING VALUE CHECK")
print("=" * 70)

check_columns = [
    "latitude_normalized",
    "longitude_normalized",
    "latitude_centered",
    "longitude_centered",
    "spatial_distance_from_center",
    "chlorophyll_a_log1p"
]

print(
    df[check_columns]
    .isna()
    .sum()
)


# ======================================================================
# ROW COUNT CHECK
# ======================================================================

print("\n" + "=" * 70)
print("ROW COUNT CHECK")
print("=" * 70)

print(
    "Rows:",
    len(df)
)


# ======================================================================
# FINAL DATASET STRUCTURE
# ======================================================================

print("\n" + "=" * 70)
print("FINAL DATASET STRUCTURE")
print("=" * 70)

print(
    "Rows:",
    len(df)
)

print(
    "Columns:",
    len(df.columns)
)

print("\nNew spatial features:")
print("✓ latitude_normalized")
print("✓ longitude_normalized")
print("✓ latitude_centered")
print("✓ longitude_centered")
print("✓ spatial_distance_from_center")

print("\nNew CHL transformation:")
print("✓ chlorophyll_a_log1p")


# ======================================================================
# SAVE DATASET
# ======================================================================

df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ======================================================================
# SAVE COMPARISON TABLE
# ======================================================================

COMPARISON_FILE = (
    OUTPUT_DIR
    / "chlorophyll_a_transformation_comparison.csv"
)

comparison.to_csv(
    COMPARISON_FILE,
    index=False
)


# ======================================================================
# COMPLETE
# ======================================================================

print("\n" + "=" * 70)
print("STEP 31.9P-6 COMPLETE")
print("=" * 70)

print("\nFinal dataset shape:")
print(df.shape)

print("\nDataset saved to:")
print(OUTPUT_FILE)

print("\nCHL transformation comparison saved to:")
print(COMPARISON_FILE)

print("\nFigures saved:")
print(original_plot)
print(log_plot)
```

```python
# ================================================================
# STEP 31.9P-8 — TEMPORAL TRAIN / VALIDATION / TEST SPLIT
# ================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 31.9P-8 — TEMPORAL TRAIN / VALIDATION / TEST SPLIT")
print("=" * 70)


# ------------------------------------------------
# PATHS
# ------------------------------------------------

BASE_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

FEATURE_DIR = BASE_DIR / "04_feature_engineering"

INPUT_FILE = (
    FEATURE_DIR /
    "final_daily_modelling_dataset.csv"
)

SPLIT_REPORT_FILE = (
    FEATURE_DIR /
    "temporal_train_validation_test_split_report.csv"
)

SPLIT_BOUNDARIES_FILE = (
    FEATURE_DIR /
    "temporal_split_boundaries.txt"
)


# ------------------------------------------------
# LOAD DATA
# ------------------------------------------------

print("\n" + "=" * 70)
print("LOADING FINAL MODELLING DATASET")
print("=" * 70)

df = pd.read_csv(
    INPUT_FILE,
    parse_dates=["date"]
)

print("Shape:", df.shape)

print(
    "Date range:",
    df["date"].min(),
    "to",
    df["date"].max()
)

print(
    "Number of spatial locations:",
    df[["latitude", "longitude"]]
    .drop_duplicates()
    .shape[0]
)


# ------------------------------------------------
# SORT DATA
# ------------------------------------------------

print("\n" + "=" * 70)
print("SORTING CHRONOLOGICALLY")
print("=" * 70)

df = df.sort_values(
    ["date", "latitude", "longitude"]
).reset_index(drop=True)

print("✓ Dataset sorted by date and spatial location.")


# ------------------------------------------------
# UNIQUE DATES
# ------------------------------------------------

unique_dates = np.sort(
    df["date"].dropna().unique()
)

number_of_dates = len(unique_dates)

print("\n" + "=" * 70)
print("UNIQUE DATE INFORMATION")
print("=" * 70)

print("Number of unique dates:", number_of_dates)

print("First date:", unique_dates[0])
print("Last date:", unique_dates[-1])


# ------------------------------------------------
# 70 / 15 / 15 CHRONOLOGICAL SPLIT
# ------------------------------------------------

train_date_count = int(
    number_of_dates * 0.70
)

validation_date_count = int(
    number_of_dates * 0.15
)

train_end_date = (
    unique_dates[train_date_count - 1]
)

validation_end_date = (
    unique_dates[
        train_date_count +
        validation_date_count - 1
    ]
)

test_start_date = (
    unique_dates[
        train_date_count +
        validation_date_count
    ]
)


# ------------------------------------------------
# ASSIGN SPLIT
# ------------------------------------------------

def assign_split(date):

    if date <= train_end_date:
        return "train"

    elif date <= validation_end_date:
        return "validation"

    else:
        return "test"


df["data_split"] = df["date"].apply(assign_split)


# ------------------------------------------------
# SPLIT SUMMARY
# ------------------------------------------------

print("\n" + "=" * 70)
print("TEMPORAL SPLIT BOUNDARIES")
print("=" * 70)

print(
    "TRAIN:",
    df.loc[
        df["data_split"] == "train",
        "date"
    ].min(),
    "→",
    df.loc[
        df["data_split"] == "train",
        "date"
    ].max()
)

print(
    "VALIDATION:",
    df.loc[
        df["data_split"] == "validation",
        "date"
    ].min(),
    "→",
    df.loc[
        df["data_split"] == "validation",
        "date"
    ].max()
)

print(
    "TEST:",
    df.loc[
        df["data_split"] == "test",
        "date"
    ].min(),
    "→",
    df.loc[
        df["data_split"] == "test",
        "date"
    ].max()
)


# ------------------------------------------------
# ROW COUNTS
# ------------------------------------------------

print("\n" + "=" * 70)
print("ROWS PER SPLIT")
print("=" * 70)

split_counts = (
    df["data_split"]
    .value_counts()
    .reindex(
        ["train", "validation", "test"]
    )
)

print(split_counts)

print("\nPercentages:")

split_percentages = (
    df["data_split"]
    .value_counts(normalize=True)
    .reindex(
        ["train", "validation", "test"]
    ) * 100
)

print(split_percentages)


# ------------------------------------------------
# UNIQUE DATE COUNTS
# ------------------------------------------------

print("\n" + "=" * 70)
print("UNIQUE DATES PER SPLIT")
print("=" * 70)

for split_name in [
    "train",
    "validation",
    "test"
]:

    split_df = df[
        df["data_split"] == split_name
    ]

    print(
        f"{split_name:12s}: "
        f"{split_df['date'].nunique()} dates"
    )


# ------------------------------------------------
# SPATIAL LOCATION CHECK
# ------------------------------------------------

print("\n" + "=" * 70)
print("SPATIAL LOCATION CHECK")
print("=" * 70)

for split_name in [
    "train",
    "validation",
    "test"
]:

    split_df = df[
        df["data_split"] == split_name
    ]

    locations = (
        split_df[
            ["latitude", "longitude"]
        ]
        .drop_duplicates()
        .shape[0]
    )

    print(
        f"{split_name:12s}: "
        f"{locations} spatial locations"
    )


# ------------------------------------------------
# CHL TARGET STATISTICS BY SPLIT
# ------------------------------------------------

print("\n" + "=" * 70)
print("CHLOROPHYLL-a TARGET DISTRIBUTION BY SPLIT")
print("=" * 70)

target_summary = (
    df.groupby("data_split")["chlorophyll_a"]
    .agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
    .reindex(
        ["train", "validation", "test"]
    )
)

print(target_summary)


# ------------------------------------------------
# LOG TARGET STATISTICS
# ------------------------------------------------

print("\n" + "=" * 70)
print("LOG-TRANSFORMED CHL TARGET BY SPLIT")
print("=" * 70)

log_target_summary = (
    df.groupby("data_split")[
        "chlorophyll_a_log1p"
    ]
    .agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
    .reindex(
        ["train", "validation", "test"]
    )
)

print(log_target_summary)


# ------------------------------------------------
# TEMPORAL ORDER VALIDATION
# ------------------------------------------------

print("\n" + "=" * 70)
print("TEMPORAL ORDER VALIDATION")
print("=" * 70)

train_max = df.loc[
    df["data_split"] == "train",
    "date"
].max()

validation_min = df.loc[
    df["data_split"] == "validation",
    "date"
].min()

validation_max = df.loc[
    df["data_split"] == "validation",
    "date"
].max()

test_min = df.loc[
    df["data_split"] == "test",
    "date"
].min()


if train_max < validation_min:
    print(
        "✓ Training ends before validation begins."
    )
else:
    print(
        "⚠️ Training/validation temporal overlap detected."
    )


if validation_max < test_min:
    print(
        "✓ Validation ends before testing begins."
    )
else:
    print(
        "⚠️ Validation/test temporal overlap detected."
    )


# ------------------------------------------------
# CHECK FOR DATE OVERLAP
# ------------------------------------------------

train_dates = set(
    df.loc[
        df["data_split"] == "train",
        "date"
    ]
)

validation_dates = set(
    df.loc[
        df["data_split"] == "validation",
        "date"
    ]
)

test_dates = set(
    df.loc[
        df["data_split"] == "test",
        "date"
    ]
)

print("\nDate overlap checks:")

print(
    "Train ∩ Validation:",
    len(train_dates & validation_dates)
)

print(
    "Validation ∩ Test:",
    len(validation_dates & test_dates)
)

print(
    "Train ∩ Test:",
    len(train_dates & test_dates)
)


# ------------------------------------------------
# SAVE SPLIT REPORT
# ------------------------------------------------

split_report = pd.DataFrame({
    "split": [
        "train",
        "validation",
        "test"
    ],

    "start_date": [
        df.loc[
            df["data_split"] == "train",
            "date"
        ].min(),

        df.loc[
            df["data_split"] == "validation",
            "date"
        ].min(),

        df.loc[
            df["data_split"] == "test",
            "date"
        ].min()
    ],

    "end_date": [
        df.loc[
            df["data_split"] == "train",
            "date"
        ].max(),

        df.loc[
            df["data_split"] == "validation",
            "date"
        ].max(),

        df.loc[
            df["data_split"] == "test",
            "date"
        ].max()
    ],

    "rows": [
        len(
            df[
                df["data_split"] == "train"
            ]
        ),

        len(
            df[
                df["data_split"] == "validation"
            ]
        ),

        len(
            df[
                df["data_split"] == "test"
            ]
        )
    ],

    "unique_dates": [
        df.loc[
            df["data_split"] == "train",
            "date"
        ].nunique(),

        df.loc[
            df["data_split"] == "validation",
            "date"
        ].nunique(),

        df.loc[
            df["data_split"] == "test",
            "date"
        ].nunique()
    ],

    "spatial_locations": [
        df.loc[
            df["data_split"] == "train",
            ["latitude", "longitude"]
        ].drop_duplicates().shape[0],

        df.loc[
            df["data_split"] == "validation",
            ["latitude", "longitude"]
        ].drop_duplicates().shape[0],

        df.loc[
            df["data_split"] == "test",
            ["latitude", "longitude"]
        ].drop_duplicates().shape[0]
    ]
})

split_report.to_csv(
    SPLIT_REPORT_FILE,
    index=False
)


# ------------------------------------------------
# SAVE BOUNDARIES
# ------------------------------------------------

with open(
    SPLIT_BOUNDARIES_FILE,
    "w"
) as f:

    f.write(
        "TEMPORAL TRAIN / VALIDATION / TEST SPLIT\n"
    )

    f.write(
        "=========================================\n\n"
    )

    f.write(
        f"Training:\n"
        f"{df.loc[df['data_split']=='train','date'].min()}"
        f" -> "
        f"{df.loc[df['data_split']=='train','date'].max()}\n\n"
    )

    f.write(
        f"Validation:\n"
        f"{df.loc[df['data_split']=='validation','date'].min()}"
        f" -> "
        f"{df.loc[df['data_split']=='validation','date'].max()}\n\n"
    )

    f.write(
        f"Testing:\n"
        f"{df.loc[df['data_split']=='test','date'].min()}"
        f" -> "
        f"{df.loc[df['data_split']=='test','date'].max()}\n"
    )


# ------------------------------------------------
# SAVE DATASET WITH SPLIT LABEL
# ------------------------------------------------

SPLIT_DATASET_FILE = (
    FEATURE_DIR /
    "final_daily_modelling_dataset_with_split.csv"
)

df.to_csv(
    SPLIT_DATASET_FILE,
    index=False
)


# ------------------------------------------------
# FINAL OUTPUT
# ------------------------------------------------

print("\n" + "=" * 70)
print("STEP 31.9P-8 COMPLETE")
print("=" * 70)

print("\nSplit report saved to:")
print(SPLIT_REPORT_FILE)

print("\nSplit boundaries saved to:")
print(SPLIT_BOUNDARIES_FILE)

print("\nDataset with split labels saved to:")
print(SPLIT_DATASET_FILE)

print("\nNext step:")
print("STEP 31.9P-9 — TRAIN-ONLY PREPROCESSING & FEATURE SCALING")
```

```python
# ======================================================================
# STEP 31.9P-9 — TRAIN-ONLY PREPROCESSING & FEATURE SCALING
# ======================================================================

from google.colab import drive
drive.mount("/content/drive")

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import joblib

print("=" * 70)
print("STEP 31.9P-9 — TRAIN-ONLY PREPROCESSING & FEATURE SCALING")
print("=" * 70)

# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/Indhi_Marine_Research_Data")

INPUT_FILE = (
    BASE_DIR
    / "04_feature_engineering"
    / "final_daily_modelling_dataset_with_split.csv"
)

OUTPUT_DIR = BASE_DIR / "05_model_preprocessing"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("\nInput file:")
print(INPUT_FILE)

# ----------------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------------

df = pd.read_csv(
    INPUT_FILE,
    parse_dates=["date"]
)

print("\n" + "=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Shape:", df.shape)
print("Date range:", df["date"].min(), "to", df["date"].max())
print("Spatial locations:",
      df[["latitude", "longitude"]].drop_duplicates().shape[0])

print("\nSplit distribution:")
print(df["data_split"].value_counts())

# ----------------------------------------------------------------------
# SORT
# ----------------------------------------------------------------------

df = df.sort_values(
    ["date", "latitude", "longitude"]
).reset_index(drop=True)

# ----------------------------------------------------------------------
# IDENTIFY TARGET / METADATA
# ----------------------------------------------------------------------

TARGET = "chlorophyll_a"
LOG_TARGET = "chlorophyll_a_log1p"
SPLIT_COLUMN = "data_split"

# These are identifiers / non-model metadata.
# They are retained in the output but are NOT scaled as predictors.
NON_FEATURE_COLUMNS = [
    "date",
    "latitude",
    "longitude",
    TARGET,
    LOG_TARGET,
    SPLIT_COLUMN,
    "season"
]

# Numeric candidate features
numeric_columns = df.select_dtypes(
    include=[np.number]
).columns.tolist()

FEATURE_COLUMNS = [
    col
    for col in numeric_columns
    if col not in NON_FEATURE_COLUMNS
]

print("\n" + "=" * 70)
print("FEATURE IDENTIFICATION")
print("=" * 70)

print("Target:", TARGET)
print("Log target:", LOG_TARGET)
print("Number of numeric predictor features:", len(FEATURE_COLUMNS))

print("\nPredictor features:")

for feature in FEATURE_COLUMNS:
    print("✓", feature)

# ----------------------------------------------------------------------
# SPLIT DATA
# ----------------------------------------------------------------------

train_df = df[df[SPLIT_COLUMN] == "train"].copy()
val_df = df[df[SPLIT_COLUMN] == "validation"].copy()
test_df = df[df[SPLIT_COLUMN] == "test"].copy()

print("\n" + "=" * 70)
print("DATA SPLITS")
print("=" * 70)

print("Training rows   :", len(train_df))
print("Validation rows :", len(val_df))
print("Testing rows    :", len(test_df))

# ----------------------------------------------------------------------
# CHECK FEATURE MISSING VALUES BEFORE IMPUTATION
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUES BEFORE PREPROCESSING")
print("=" * 70)

missing_before = pd.DataFrame({
    "train_missing": train_df[FEATURE_COLUMNS].isna().sum(),
    "validation_missing": val_df[FEATURE_COLUMNS].isna().sum(),
    "test_missing": test_df[FEATURE_COLUMNS].isna().sum()
})

missing_before["train_missing_pct"] = (
    missing_before["train_missing"]
    / len(train_df)
    * 100
)

missing_before["validation_missing_pct"] = (
    missing_before["validation_missing"]
    / len(val_df)
    * 100
)

missing_before["test_missing_pct"] = (
    missing_before["test_missing"]
    / len(test_df)
    * 100
)

print(missing_before)

# ----------------------------------------------------------------------
# TRAIN-ONLY MEDIAN IMPUTATION
# ----------------------------------------------------------------------
#
# IMPORTANT:
# The imputer is FIT ONLY on the training dataset.
#
# Validation and test data are transformed using the training medians.
#
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TRAIN-ONLY MEDIAN IMPUTATION")
print("=" * 70)

imputer = SimpleImputer(
    strategy="median"
)

# FIT ONLY ON TRAINING DATA
X_train_imputed = imputer.fit_transform(
    train_df[FEATURE_COLUMNS]
)

# APPLY TRAIN-LEARNED MEDIANS
X_val_imputed = imputer.transform(
    val_df[FEATURE_COLUMNS]
)

X_test_imputed = imputer.transform(
    test_df[FEATURE_COLUMNS]
)

print("✓ Imputer fitted using TRAINING data only.")
print("✓ Training medians applied to validation data.")
print("✓ Training medians applied to test data.")

# ----------------------------------------------------------------------
# TRAIN-ONLY STANDARD SCALING
# ----------------------------------------------------------------------
#
# StandardScaler learns:
#
# mean
# standard deviation
#
# ONLY from the training set.
#
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TRAIN-ONLY STANDARD SCALING")
print("=" * 70)

feature_scaler = StandardScaler()

# FIT ONLY ON TRAINING DATA
X_train_scaled = feature_scaler.fit_transform(
    X_train_imputed
)

# APPLY TRAIN-LEARNED SCALING
X_val_scaled = feature_scaler.transform(
    X_val_imputed
)

X_test_scaled = feature_scaler.transform(
    X_test_imputed
)

print("✓ Feature scaler fitted using TRAINING data only.")
print("✓ Training scaler applied to validation data.")
print("✓ Training scaler applied to test data.")

# ----------------------------------------------------------------------
# TARGET PREPARATION
# ----------------------------------------------------------------------
#
# Keep original chlorophyll_a unchanged.
#
# chlorophyll_a_log1p is used as the transformed target for models
# that benefit from reduced right-skew.
#
# We DO NOT fit the feature scaler using the target.
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET PREPARATION")
print("=" * 70)

y_train_raw = train_df[TARGET].to_numpy()
y_val_raw = val_df[TARGET].to_numpy()
y_test_raw = test_df[TARGET].to_numpy()

y_train_log = train_df[LOG_TARGET].to_numpy()
y_val_log = val_df[LOG_TARGET].to_numpy()
y_test_log = test_df[LOG_TARGET].to_numpy()

print("Original CHL target preserved.")
print("Log1p CHL target available.")

print("\nOriginal CHL ranges:")
print(
    "Train      :",
    y_train_raw.min(),
    "to",
    y_train_raw.max()
)

print(
    "Validation :",
    y_val_raw.min(),
    "to",
    y_val_raw.max()
)

print(
    "Test       :",
    y_test_raw.min(),
    "to",
    y_test_raw.max()
)

# ----------------------------------------------------------------------
# CREATE SCALED DATAFRAMES
# ----------------------------------------------------------------------

train_scaled_df = train_df[
    [
        "date",
        "latitude",
        "longitude",
        TARGET,
        LOG_TARGET
    ]
].copy()

val_scaled_df = val_df[
    [
        "date",
        "latitude",
        "longitude",
        TARGET,
        LOG_TARGET
    ]
].copy()

test_scaled_df = test_df[
    [
        "date",
        "latitude",
        "longitude",
        TARGET,
        LOG_TARGET
    ]
].copy()

# Add scaled features
for i, feature in enumerate(FEATURE_COLUMNS):

    train_scaled_df[feature] = X_train_scaled[:, i]
    val_scaled_df[feature] = X_val_scaled[:, i]
    test_scaled_df[feature] = X_test_scaled[:, i]

# ----------------------------------------------------------------------
# ADD SPLIT LABEL
# ----------------------------------------------------------------------

train_scaled_df[SPLIT_COLUMN] = "train"
val_scaled_df[SPLIT_COLUMN] = "validation"
test_scaled_df[SPLIT_COLUMN] = "test"

# ----------------------------------------------------------------------
# SORT OUTPUT
# ----------------------------------------------------------------------

train_scaled_df = train_scaled_df.sort_values(
    ["date", "latitude", "longitude"]
).reset_index(drop=True)

val_scaled_df = val_scaled_df.sort_values(
    ["date", "latitude", "longitude"]
).reset_index(drop=True)

test_scaled_df = test_scaled_df.sort_values(
    ["date", "latitude", "longitude"]
).reset_index(drop=True)

# ----------------------------------------------------------------------
# POST-PROCESSING MISSING VALUE CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUES AFTER IMPUTATION")
print("=" * 70)

print(
    "Train:",
    train_scaled_df[FEATURE_COLUMNS].isna().sum().sum()
)

print(
    "Validation:",
    val_scaled_df[FEATURE_COLUMNS].isna().sum().sum()
)

print(
    "Test:",
    test_scaled_df[FEATURE_COLUMNS].isna().sum().sum()
)

# ----------------------------------------------------------------------
# SCALING CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TRAINING FEATURE SCALING CHECK")
print("=" * 70)

train_means = train_scaled_df[FEATURE_COLUMNS].mean()
train_stds = train_scaled_df[FEATURE_COLUMNS].std()

print(
    "Maximum absolute training mean:",
    train_means.abs().max()
)

print(
    "Minimum training standard deviation:",
    train_stds.min()
)

print(
    "Maximum training standard deviation:",
    train_stds.max()
)

# ----------------------------------------------------------------------
# VALIDATION / TEST TRANSFORMATION CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDATION / TEST SCALING CHECK")
print("=" * 70)

print("\nValidation scaled feature mean range:")
print(
    val_scaled_df[FEATURE_COLUMNS]
    .mean()
    .min(),
    "to",
    val_scaled_df[FEATURE_COLUMNS]
    .mean()
    .max()
)

print("\nTest scaled feature mean range:")
print(
    test_scaled_df[FEATURE_COLUMNS]
    .mean()
    .min(),
    "to",
    test_scaled_df[FEATURE_COLUMNS]
    .mean()
    .max()
)

# ----------------------------------------------------------------------
# SAVE DATASETS
# ----------------------------------------------------------------------

TRAIN_FILE = OUTPUT_DIR / "train_scaled.csv"
VAL_FILE = OUTPUT_DIR / "validation_scaled.csv"
TEST_FILE = OUTPUT_DIR / "test_scaled.csv"

train_scaled_df.to_csv(
    TRAIN_FILE,
    index=False
)

val_scaled_df.to_csv(
    VAL_FILE,
    index=False
)

test_scaled_df.to_csv(
    TEST_FILE,
    index=False
)

# ----------------------------------------------------------------------
# SAVE PREPROCESSING OBJECTS
# ----------------------------------------------------------------------

IMPUTER_FILE = OUTPUT_DIR / "feature_median_imputer.joblib"
SCALER_FILE = OUTPUT_DIR / "feature_standard_scaler.joblib"

joblib.dump(
    imputer,
    IMPUTER_FILE
)

joblib.dump(
    feature_scaler,
    SCALER_FILE
)

# ----------------------------------------------------------------------
# SAVE FEATURE LIST
# ----------------------------------------------------------------------

FEATURE_LIST_FILE = OUTPUT_DIR / "model_feature_columns.csv"

pd.DataFrame({
    "feature": FEATURE_COLUMNS
}).to_csv(
    FEATURE_LIST_FILE,
    index=False
)

# ----------------------------------------------------------------------
# SAVE PREPROCESSING SUMMARY
# ----------------------------------------------------------------------

summary = pd.DataFrame({
    "item": [
        "input_rows",
        "number_of_features",
        "train_rows",
        "validation_rows",
        "test_rows",
        "imputation_strategy",
        "scaling_method",
        "scaler_fitted_on",
        "target",
        "log_target"
    ],
    "value": [
        len(df),
        len(FEATURE_COLUMNS),
        len(train_scaled_df),
        len(val_scaled_df),
        len(test_scaled_df),
        "median",
        "StandardScaler",
        "training data only",
        TARGET,
        LOG_TARGET
    ]
})

SUMMARY_FILE = OUTPUT_DIR / "preprocessing_summary.csv"

summary.to_csv(
    SUMMARY_FILE,
    index=False
)

# ----------------------------------------------------------------------
# FINAL VERIFICATION
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL PREPROCESSING VERIFICATION")
print("=" * 70)

print("Train shape      :", train_scaled_df.shape)
print("Validation shape :", val_scaled_df.shape)
print("Test shape       :", test_scaled_df.shape)

print("\nTrain date range:")
print(
    train_scaled_df["date"].min(),
    "→",
    train_scaled_df["date"].max()
)

print("\nValidation date range:")
print(
    val_scaled_df["date"].min(),
    "→",
    val_scaled_df["date"].max()
)

print("\nTest date range:")
print(
    test_scaled_df["date"].min(),
    "→",
    test_scaled_df["date"].max()
)

print("\n" + "=" * 70)
print("STEP 31.9P-9 COMPLETE")
print("=" * 70)

print("\nSaved files:")

print(TRAIN_FILE)
print(VAL_FILE)
print(TEST_FILE)
print(IMPUTER_FILE)
print(SCALER_FILE)
print(FEATURE_LIST_FILE)
print(SUMMARY_FILE)

print("\n✓ Imputation fitted on TRAIN only.")
print("✓ Scaling fitted on TRAIN only.")
print("✓ Validation transformed using TRAIN parameters.")
print("✓ Test transformed using TRAIN parameters.")
print("✓ Original chlorophyll_a target preserved.")
print("✓ Log1p chlorophyll_a target preserved.")
```

```python
# ======================================================================
# STEP 31.9P-10 PART C — FEATURE REDUNDANCY INTERPRETATION
# ======================================================================

from google.colab import drive
drive.mount("/content/drive")

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 31.9P-10 PART C — FEATURE REDUNDANCY INTERPRETATION")
print("=" * 70)

# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

PREPROCESS_DIR = BASE_DIR / "05_model_preprocessing"

CORR_FILE = (
    PREPROCESS_DIR /
    "training_feature_correlation_matrix.csv"
)

FEATURE_FILE = (
    PREPROCESS_DIR /
    "feature_inventory_and_categories.csv"
)

TARGET_CORR_FILE = (
    PREPROCESS_DIR /
    "training_feature_target_correlations.csv"
)

OUTPUT_DIR = PREPROCESS_DIR

# ----------------------------------------------------------------------
# LOAD FILES
# ----------------------------------------------------------------------

corr_matrix = pd.read_csv(
    CORR_FILE,
    index_col=0
)

feature_inventory = pd.read_csv(
    FEATURE_FILE
)

target_corr = pd.read_csv(
    TARGET_CORR_FILE
)

print("\n" + "=" * 70)
print("FILES LOADED")
print("=" * 70)

print("Correlation matrix shape:", corr_matrix.shape)
print(
    "Feature inventory shape:",
    feature_inventory.shape
)
print(
    "Target correlation rows:",
    len(target_corr)
)

# ----------------------------------------------------------------------
# CREATE CORRELATION PAIRS
# ----------------------------------------------------------------------

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

pairs.columns = [
    "feature_1",
    "feature_2",
    "pearson_r"
]

pairs["absolute_r"] = (
    pairs["pearson_r"].abs()
)

pairs = pairs.sort_values(
    "absolute_r",
    ascending=False
).reset_index(drop=True)

# ----------------------------------------------------------------------
# MERGE CATEGORY INFORMATION
# ----------------------------------------------------------------------

category_map = dict(
    zip(
        feature_inventory["feature"],
        feature_inventory["category"]
    )
)

pairs["category_1"] = pairs[
    "feature_1"
].map(category_map)

pairs["category_2"] = pairs[
    "feature_2"
].map(category_map)

# ----------------------------------------------------------------------
# CATEGORY-LEVEL REDUNDANCY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("HIGH CORRELATION BY FEATURE CATEGORY")
print("=" * 70)

category_summary = []

for category in sorted(
    feature_inventory["category"].unique()
):

    category_features = [
        f for f, c in category_map.items()
        if c == category
    ]

    category_pairs = pairs[
        pairs["feature_1"].isin(category_features)
        &
        pairs["feature_2"].isin(category_features)
    ]

    high_90 = (
        category_pairs["absolute_r"] >= 0.90
    ).sum()

    high_95 = (
        category_pairs["absolute_r"] >= 0.95
    ).sum()

    high_99 = (
        category_pairs["absolute_r"] >= 0.99
    ).sum()

    category_summary.append({
        "category": category,
        "number_of_features": len(
            category_features
        ),
        "pairs_above_0.90": int(high_90),
        "pairs_above_0.95": int(high_95),
        "pairs_above_0.99": int(high_99)
    })

category_summary_df = pd.DataFrame(
    category_summary
)

print(
    category_summary_df.to_string(
        index=False
    )
)

# ----------------------------------------------------------------------
# IDENTICAL / NEAR-IDENTICAL SPATIAL FEATURES
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("SPATIAL REDUNDANCY CHECK")
print("=" * 70)

spatial_features = [
    f for f, c in category_map.items()
    if c == "Spatial"
]

spatial_pairs = pairs[
    pairs["feature_1"].isin(spatial_features)
    &
    pairs["feature_2"].isin(spatial_features)
].copy()

print(
    spatial_pairs.to_string(
        index=False
    )
)

# ----------------------------------------------------------------------
# TEMPORAL REDUNDANCY CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TEMPORAL REDUNDANCY CHECK")
print("=" * 70)

temporal_features = [
    f for f, c in category_map.items()
    if c == "Temporal"
]

temporal_pairs = pairs[
    pairs["feature_1"].isin(temporal_features)
    &
    pairs["feature_2"].isin(temporal_features)
].copy()

print(
    temporal_pairs.head(30).to_string(
        index=False
    )
)

# ----------------------------------------------------------------------
# CHL TEMPORAL FEATURE REDUNDANCY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("CHL TEMPORAL FEATURE REDUNDANCY")
print("=" * 70)

chl_temporal_categories = [
    "CHL lag",
    "CHL rolling"
]

chl_temporal_features = [
    f for f, c in category_map.items()
    if c in chl_temporal_categories
]

chl_temporal_pairs = pairs[
    pairs["feature_1"].isin(
        chl_temporal_features
    )
    &
    pairs["feature_2"].isin(
        chl_temporal_features
    )
].copy()

print(
    chl_temporal_pairs.head(40).to_string(
        index=False
    )
)

# ----------------------------------------------------------------------
# ENVIRONMENTAL REDUNDANCY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("ENVIRONMENTAL FEATURE REDUNDANCY")
print("=" * 70)

environmental_categories = [
    "Current environmental",
    "Environmental lag",
    "Environmental rolling"
]

environmental_features = [
    f for f, c in category_map.items()
    if c in environmental_categories
]

environmental_pairs = pairs[
    pairs["feature_1"].isin(
        environmental_features
    )
    &
    pairs["feature_2"].isin(
        environmental_features
    )
].copy()

print(
    environmental_pairs.head(40).to_string(
        index=False
    )
)

# ----------------------------------------------------------------------
# TARGET CORRELATION SUMMARY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP FEATURES ASSOCIATED WITH CHL TARGET")
print("=" * 70)

target_summary = target_corr.copy()

target_summary["category"] = (
    target_summary["feature"].map(
        category_map
    )
)

print(
    target_summary.head(30).to_string(
        index=False
    )
)

# ----------------------------------------------------------------------
# SAVE CATEGORY SUMMARY
# ----------------------------------------------------------------------

CATEGORY_FILE = (
    OUTPUT_DIR /
    "feature_redundancy_category_summary.csv"
)

category_summary_df.to_csv(
    CATEGORY_FILE,
    index=False
)

# ----------------------------------------------------------------------
# SAVE CHL TEMPORAL REDUNDANCY
# ----------------------------------------------------------------------

CHL_REDUNDANCY_FILE = (
    OUTPUT_DIR /
    "chl_temporal_feature_redundancy.csv"
)

chl_temporal_pairs.to_csv(
    CHL_REDUNDANCY_FILE,
    index=False
)

# ----------------------------------------------------------------------
# SAVE ENVIRONMENTAL REDUNDANCY
# ----------------------------------------------------------------------

ENV_REDUNDANCY_FILE = (
    OUTPUT_DIR /
    "environmental_feature_redundancy.csv"
)

environmental_pairs.to_csv(
    ENV_REDUNDANCY_FILE,
    index=False
)

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 31.9P-10 PART C COMPLETE")
print("=" * 70)

print("\nFiles saved:")

print(CATEGORY_FILE)
print(CHL_REDUNDANCY_FILE)
print(ENV_REDUNDANCY_FILE)

print("\nNo features have been deleted at this stage.")

print(
    "\nNext:"
)

print(
    "STEP 31.9P-10 PART D — FINAL FEATURE GROUP DECISION"
)
```

```python
# ======================================================================
# STEP 31.9P-10 PART D — FINAL FEATURE GROUP DECISION
# ======================================================================

from google.colab import drive
drive.mount("/content/drive")

import pandas as pd
from pathlib import Path

print("=" * 70)
print("STEP 31.9P-10 PART D — FINAL FEATURE GROUP DECISION")
print("=" * 70)

# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE_DIR = Path(
    "/content/drive/MyDrive/Indhi_Marine_Research_Data"
)

PREPROCESS_DIR = BASE_DIR / "05_model_preprocessing"

FEATURE_FILE = (
    PREPROCESS_DIR /
    "model_feature_columns.csv"
)

OUTPUT_DIR = PREPROCESS_DIR

# ----------------------------------------------------------------------
# LOAD ORIGINAL FEATURE LIST
# ----------------------------------------------------------------------

feature_df = pd.read_csv(FEATURE_FILE)

ALL_FEATURES = feature_df["feature"].tolist()

print("\nTotal original predictors:", len(ALL_FEATURES))

# ======================================================================
# GROUP A — FULL FEATURE SET
# ======================================================================

GROUP_A = ALL_FEATURES.copy()

# ======================================================================
# GROUP B — REDUCED FEATURE SET
# ======================================================================
#
# Remove obvious redundant representations:
#
# latitude_normalized  == latitude_centered
# longitude_normalized == longitude_centered
#
# Keep normalized spatial representations.
#
# For calendar information, retain the useful cyclic representation
# and remove redundant raw counters.
# ======================================================================

REMOVE_GROUP_B = [
    "latitude_centered",
    "longitude_centered",

    "day",
    "day_of_year",
    "week_of_year"
]

GROUP_B = [
    feature
    for feature in ALL_FEATURES
    if feature not in REMOVE_GROUP_B
]

# ======================================================================
# GROUP C — CORE FORECASTING FEATURE SET
# ======================================================================
#
# Focus on:
#   - CHL history
#   - current environmental conditions
#   - selected environmental lags
#   - environmental rolling history
#   - seasonality
#   - spatial context
# ======================================================================

CORE_CHL = [
    "chl_lag_1",
    "chl_lag_2",
    "chl_lag_3",
    "chl_lag_7",
    "chl_lag_14",
    "chl_lag_30",

    "chl_rolling_mean_3",
    "chl_rolling_mean_7",
    "chl_rolling_mean_14",

    "chl_rolling_std_7",
    "chl_rolling_std_14"
]

CORE_ENVIRONMENTAL = [
    "sst",
    "sss",
    "nitrate",
    "phosphate",
    "silicate"
]

CORE_ENVIRONMENTAL_LAGS = [
    "sst_lag_1",
    "sst_lag_7",
    "sst_lag_14",

    "sss_lag_1",
    "sss_lag_7",
    "sss_lag_14",

    "nitrate_lag_1",
    "nitrate_lag_7",
    "nitrate_lag_14",

    "phosphate_lag_1",
    "phosphate_lag_7",
    "phosphate_lag_14",

    "silicate_lag_1",
    "silicate_lag_7",
    "silicate_lag_14"
]

CORE_ENVIRONMENTAL_ROLLING = [
    "sst_rolling_mean_7",
    "sst_rolling_mean_14",

    "sss_rolling_mean_7",
    "sss_rolling_mean_14",

    "nitrate_rolling_mean_7",
    "nitrate_rolling_mean_14",

    "phosphate_rolling_mean_7",
    "phosphate_rolling_mean_14",

    "silicate_rolling_mean_7",
    "silicate_rolling_mean_14"
]

CORE_TEMPORAL = [
    "year",
    "month",

    "day_of_year_sin",
    "day_of_year_cos",

    "month_sin",
    "month_cos",

    "day_of_week_sin",
    "day_of_week_cos"
]

CORE_SPATIAL = [
    "latitude_normalized",
    "longitude_normalized",
    "spatial_distance_from_center"
]

GROUP_C = (
    CORE_CHL
    + CORE_ENVIRONMENTAL
    + CORE_ENVIRONMENTAL_LAGS
    + CORE_ENVIRONMENTAL_ROLLING
    + CORE_TEMPORAL
    + CORE_SPATIAL
)

# Remove accidental duplicates while preserving order
GROUP_C = list(dict.fromkeys(GROUP_C))

# ----------------------------------------------------------------------
# VERIFY FEATURE EXISTENCE
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE EXISTENCE CHECK")
print("=" * 70)

for group_name, group_features in [
    ("GROUP A", GROUP_A),
    ("GROUP B", GROUP_B),
    ("GROUP C", GROUP_C)
]:

    missing = [
        feature
        for feature in group_features
        if feature not in ALL_FEATURES
    ]

    print(
        f"\n{group_name}: {len(group_features)} features"
    )

    if missing:
        print("⚠ Missing features:")
        for feature in missing:
            print("   ", feature)
    else:
        print("✓ All features exist.")

# ----------------------------------------------------------------------
# GROUP SUMMARY
# ----------------------------------------------------------------------

summary = pd.DataFrame([
    {
        "feature_group": "Group_A_Full",
        "description": "All available engineered predictors",
        "feature_count": len(GROUP_A)
    },
    {
        "feature_group": "Group_B_Reduced",
        "description": (
            "Reduced redundancy while retaining "
            "temporal and environmental information"
        ),
        "feature_count": len(GROUP_B)
    },
    {
        "feature_group": "Group_C_Core",
        "description": (
            "Compact daily CHL forecasting feature set"
        ),
        "feature_count": len(GROUP_C)
    }
])

print("\n" + "=" * 70)
print("FEATURE GROUP SUMMARY")
print("=" * 70)

print(summary.to_string(index=False))

# ----------------------------------------------------------------------
# SAVE FEATURE GROUP DEFINITIONS
# ----------------------------------------------------------------------

GROUP_FILE = (
    OUTPUT_DIR /
    "candidate_feature_groups.csv"
)

group_rows = []

for group_name, group_features in [
    ("Group_A_Full", GROUP_A),
    ("Group_B_Reduced", GROUP_B),
    ("Group_C_Core", GROUP_C)
]:

    for position, feature in enumerate(
        group_features,
        start=1
    ):

        group_rows.append({
            "feature_group": group_name,
            "feature_order": position,
            "feature": feature
        })

group_df = pd.DataFrame(group_rows)

group_df.to_csv(
    GROUP_FILE,
    index=False
)

# ----------------------------------------------------------------------
# SAVE SUMMARY
# ----------------------------------------------------------------------

SUMMARY_FILE = (
    OUTPUT_DIR /
    "candidate_feature_group_summary.csv"
)

summary.to_csv(
    SUMMARY_FILE,
    index=False
)

# ----------------------------------------------------------------------
# FINAL OUTPUT
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 31.9P-10 PART D COMPLETE")
print("=" * 70)

print("\nCandidate feature groups saved to:")
print(GROUP_FILE)

print("\nFeature-group summary saved to:")
print(SUMMARY_FILE)

print("\nNo original modelling dataset was overwritten.")
print("No original features were permanently deleted.")

print("\nNext:")
print("STEP 31.9P-10 PART E — FEATURE GROUP COMPARISON")
```
